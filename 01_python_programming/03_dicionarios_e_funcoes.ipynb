````python
def simular(*respostas):
    """Devolve uma função que age como input(), mas lê de uma lista pronta."""
    it = iter(respostas)
    def entrada(msg=''):
        resp = next(it)
        print(f'{msg}{resp}')
        return resp
    return entrada
````

## 1. Dicionários

Um dicionário guarda dados em pares **chave → valor**, entre chaves `{ }`. Em vez de acessar por posição (como na lista), acessamos pelo **nome da chave**.

| | Lista | Tupla | Dicionário |
|---|---|---|---|
| Sintaxe | `[1, 2]` | `(1, 2)` | `{'a': 1, 'b': 2}` |
| Acesso | por posição | por posição | **por chave** |
| Mutável | sim | não | sim |
| Chaves repetidas | — | — | **não** (a última sobrescreve) |

````python
dici = {'nome': 'Wiliam', 'idade': 78}

print(dici['nome'])         # acessa pelo nome da chave
print(dici.keys())          # todas as chaves
print(dici.values())        # todos os valores
print(dici.items())         # pares (chave, valor)
````

### Criando, alterando e consultando
- `dici[chave] = valor` → **cria** a chave se não existir, ou **troca** o valor se já existir.
- `chave in dici` → verifica se a chave existe.
- `dici.get(chave, padrão)` → devolve o valor, ou o `padrão` (em vez de dar erro) se a chave não existir.
- `del dici[chave]` ou `dici.pop(chave)` → remove.

````python
dici = {}
dici['peso'] = 52
dici['idade'] = 100
dici['sexo'] = 'm'
print(dici)

dici['idade'] = 101                       # chave já existe: o valor é substituído
print(dici)

print('peso' in dici)                     # True
print(dici.get('altura', 'não informado'))

dici.pop('sexo')                          # remove a chave 'sexo'
print(dici)
````

### Percorrendo um dicionário

````python
dici = {'peso': 52, 'idade': 100, 'sexo': 'm'}

print('Só as chaves:')
for chave in dici.keys():
    print(' ', chave)

print('Só os valores:')
for valor in dici.values():
    print(' ', valor)

print('Chave e valor:')
for chave, valor in dici.items():
    print(' ', chave, '->', valor)
````

### Uma chave com vários valores
O valor de uma chave pode ser uma **lista**. Assim guardamos várias medidas sob o mesmo nome.

````python
entrada = simular('peso', '50', 's', '30', 's', '20', 'n')

dici = {}
chave = entrada('Digite o nome da chave: ')
dici[chave] = []                                   # a chave guarda uma lista vazia
while True:
    dici[chave].append(float(entrada('Digite o valor: ')))
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print(dici)
````

### Sem permitir chave repetida
Antes de inserir, testamos com `in` para não sobrescrever um dado existente.

````python
entrada = simular('a', '1', 's', 'b', '2', 's', 'a', '3', 'n')

dici = {}
while True:
    chave = str(entrada('Digite a chave: '))
    valor = float(entrada('Digite o valor: '))
    if chave not in dici:
        dici[chave] = valor
        print('Incluído no dicionário')
    else:
        print('Chave já existente. Dado NÃO incluído')
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print(f'Dicionário: {dici}')
print(f'Chaves:  {list(dici.keys())}')
print(f'Valores: {list(dici.values())}')
````

### Ordenando um dicionário
Um dicionário em si não tem "ordem para ordenar", mas podemos ordenar seus **itens** (`dici.items()`, uma lista de pares `(chave, valor)`) com `sorted`. O argumento `key` diz **por qual parte do par** ordenar:

- `itemgetter(1)` → pelo valor · `itemgetter(0)` → pela chave
- `lambda par: par[1]` faz a mesma coisa (veremos `lambda` mais adiante)

````python
from operator import itemgetter

dici = {'peso2': 92, 'peso4': 26, 'peso1': 55, 'peso3': 105}

ord_cre_valores   = sorted(dici.items(), key=itemgetter(1))
ord_cre_chave     = sorted(dici.items(), key=itemgetter(0))
ord_decre_valores = sorted(dici.items(), key=itemgetter(1), reverse=True)

print('Crescente por valor: ', ord_cre_valores)
print('Crescente por chave: ', ord_cre_chave)
print('Decrescente por valor:', ord_decre_valores)

# Voltando a ser um dicionário:
print(dict(ord_decre_valores))
````

````python
from random import randint
from operator import itemgetter

dici = {}
for i in range(1, 5):
    dici[f'pessoa{i}'] = randint(1, 50)

print(dici)
print('Em ordem decrescente:', sorted(dici.items(), key=itemgetter(1), reverse=True))
````

## 2. Funções

Uma função é um **bloco de código com nome**, que você escreve uma vez e usa quantas vezes quiser. Evita repetição e deixa o programa organizado.

```python
def nome_da_funcao(parametros):
    # corpo (indentado)
    return resultado       # opcional
```
- **Definir** (`def`) não executa nada; só ensina o Python a fazer aquilo.
- **Chamar** (`nome(...)`) é o que executa.
- ⚠️ A função precisa ser **definida antes de ser chamada** (a ordem das células importa).

### Função sem parâmetros

````python
def linha():
    print('-' * 20)


linha()
print('Wiliam')
linha()
````

### Função com parâmetro
O parâmetro é uma variável que recebe o valor passado na chamada.

````python
def mensagem(msg):
    print('-' * 30)
    print(msg)
    print('-' * 30)


mensagem('Curso de Python')
mensagem('Bom dia')
mensagem('Aula legal')
````

````python
def soma(a, b):
    s = a + b
    print(f'A soma de {a} + {b} = {s}')


soma(2, 3)
soma(8, 5)
soma(0, 0)
soma(13, 6)
````

### Funções que alteram uma lista
Listas são mutáveis: quando passamos uma lista para uma função e ela a modifica, **a lista original muda também**.

⚠️ Se chamarmos `triplo(lista)` **antes** do `def triplo`, o Python dá `NameError`. Defina primeiro, chame depois.

````python
def triplo(lista):
    pos = 0
    while pos < len(lista):          # índices 0, 1, 2, 3...
        lista[pos] = 3 * lista[pos]
        pos += 1
    print(f'Lista (alterada): {lista}')


lista = [2, 4, 5, 6]
print(f'Antes:  {lista}')
triplo(lista)
print(f'Depois: {lista}')            # a original foi modificada
````

### Número variável de argumentos: `*args`
Com `*num`, a função aceita **quantos argumentos você quiser**; dentro dela, `num` é uma tupla.

````python
def soma(*num):
    s = 0
    for val in num:
        s += val
    print(f'A soma dos valores vale: {s}')


soma(2, 3, 5, 6, 7)
soma(2, 3)
soma(0, 5, 6)
````

````python
def imprime_maior(mensagem, *num):
    print(mensagem, max(num))


imprime_maior('Maior:', 5, 5, 15, 1, 8)
imprime_maior('Max:', 1, 7, 9)
````

### `print` × `return`
- `print` só **mostra** o valor na tela.
- `return` **devolve** o valor para quem chamou a função, que pode guardá-lo, usá-lo em contas, etc. Ao encontrar um `return`, a função **termina na hora**.
- Uma função sem `return` devolve `None`.

````python
def soma_print(a, b):
    print(a + b)             # só mostra


def soma_return(a, b):
    return a + b             # devolve


x = soma_print(2, 3)
print('x vale:', x)          # None: a função não devolveu nada

y = soma_return(2, 3)
print('y vale:', y)          # 5
print('Dá para usar o resultado em outra conta:', y * 10)
````

````python
def soma(*num):
    s = 0
    for val in num:
        s += val
    return s


# Guardando os resultados em variáveis (cuidado: a variável NÃO pode se chamar 'soma',
# senão ela apaga a função!)
resp1 = soma(2, 3, 4)
resp2 = soma(2, 3, 5, 6, 7)
resp3 = soma(2)
print(f'As somas valem: {resp1} - {resp2} - {resp3}')

# Ou chamando direto dentro da f-string:
print(f'As somas valem: {soma(2, 3, 4)} - {soma(2, 3, 5, 6, 7)} - {soma(2)}')
````

### Devolvendo mais de um valor
Podemos devolver uma lista (ou uma tupla) e depois separar os valores.

````python
def areas_qua_tri(a, b):
    ar_qua = a * b
    ar_tri = (a * b) / 2
    return [ar_qua, ar_tri]


resp = areas_qua_tri(5, 10)
print(f'Áreas em lista: {resp}')
print(f'Área do quadrado/retângulo: {resp[0]}')
print(f'Área do triângulo: {resp[1]}')

# Também dá para "desempacotar" direto:
quadrado, triangulo = areas_qua_tri(4, 6)
print(quadrado, triangulo)
````

### Funções que devolvem `True`/`False` ou texto

````python
def eh_par(x):
    return x % 2 == 0            # a própria comparação já é True/False


def par_ou_impar(x):
    if x % 2 == 0:
        return 'par'
    else:
        return 'ímpar'


print(eh_par(4), eh_par(3))
print(par_ou_impar(4), par_ou_impar(3))
````

### Pesquisando um valor em uma lista
Retornamos a **posição** quando acha. O `return 'não encontrado'` fica **fora** do laço: só depois de testar todos.

⚠️ **Armadilha clássica:** colocar o `return 'não encontrado'` dentro de um `else` *dentro* do `for` faz a função desistir já no **primeiro** elemento diferente. Compare:

````python
def pesquise(lista, num):
    for pos, val in enumerate(lista):
        if val == num:
            return pos                      # achou: sai da função
    return 'Valor não encontrado'           # só chega aqui se NÃO achou em nenhum


lista = [10, 20, 25, 30]
print(pesquise(lista, 10))      # 0
print(pesquise(lista, 25))      # 2
print(pesquise(lista, 27))      # não encontrado
````

````python
def pesquise_errada(lista, num):
    for pos, val in enumerate(lista):
        if val == num:
            return pos
        else:
            return 'Valor não encontrado'   # ERRO: sai no 1º elemento diferente!


lista = [10, 20, 25, 30]
print(pesquise_errada(lista, 10))   # 0 (deu certo por sorte: é o 1º)
print(pesquise_errada(lista, 25))   # 'não encontrado'... mas 25 ESTÁ na lista!
````

### Parâmetros com valor padrão e argumentos nomeados
Se o parâmetro tem valor padrão, ele é **opcional** na chamada. Argumentos nomeados deixam a chamada mais legível e dispensam a ordem.

````python
def mensagem(msg, simbolo='-', tamanho=30):
    print(simbolo * tamanho)
    print(msg)
    print(simbolo * tamanho)


mensagem('Padrão')
mensagem('Com asteriscos', simbolo='*')
mensagem('Tamanho menor', tamanho=15, simbolo='=')      # a ordem não importa
````

### Docstring
O texto entre `"""` logo depois do `def` documenta a função e aparece quando se usa `help(funcao)`.

````python
def media(*notas):
    """Calcula a média aritmética das notas informadas."""
    return sum(notas) / len(notas)


print(media(7, 8.5, 6))
help(media)
````

### Escopo: variável local × global
Variáveis criadas **dentro** da função só existem lá dentro (são *locais*).

````python
x = 10                      # global


def funcao():
    x = 99                  # local: outra variável, só vale aqui dentro
    print('Dentro da função:', x)


funcao()
print('Fora da função:', x)     # continua 10

try:
    def outra():
        variavel_interna = 5
    outra()
    print(variavel_interna)     # só existia dentro de 'outra'
except NameError as erro:
    print('Erro:', erro)
````

### Função `lambda`
Uma função pequena, de uma linha, sem nome. Muito usada no `key` do `sorted`.

````python
quadrado = lambda x: x ** 2
print(quadrado(5))

dici = {'peso2': 92, 'peso4': 26, 'peso1': 55}
print(sorted(dici.items(), key=lambda par: par[1]))      # ordena pelo valor
````

### Juntando tudo: função que usa lista, laço e `return` múltiplo

````python
def estatisticas(valores):
    """Devolve (mínimo, máximo, média) de uma lista de números."""
    return min(valores), max(valores), sum(valores) / len(valores)


minimo, maximo, media_ = estatisticas([4, 8, 15, 16, 23, 42])
print(f'Mínimo: {minimo} | Máximo: {maximo} | Média: {media_:.2f}')
````

## Aprofundamento 1 — Tipagem, funções e documentação
Anotações de tipo comunicam contratos, mas **não validam** os dados automaticamente. `docstrings` descrevem comportamento, argumentos e exceções.

````python
def calcular_desconto(preco: float, percentual: float = 0.1) -> float:
    """Retorna o preço final, validando os limites do desconto."""
    if preco < 0 or not 0 <= percentual <= 1:
        raise ValueError("Preço ou percentual inválido")
    return round(preco * (1 - percentual), 2)

assert calcular_desconto(100) == 90
assert calcular_desconto(200, 0.25) == 150
print(calcular_desconto.__doc__)
````

## Aprofundamento 2 — Argumentos mutáveis e escopo
Nunca use listas ou dicionários mutáveis como valor-padrão quando se deseja um objeto novo a cada chamada.

````python
def acrescentar_item(item: str, itens: list[str] | None = None) -> list[str]:
    if itens is None:
        itens = []
    itens.append(item)
    return itens

a = acrescentar_item("A")
b = acrescentar_item("B")
assert a == ["A"] and b == ["B"]
print("Chamadas independentes:", a, b)
````

## Aprofundamento 3 — Testes simples e módulos
Prefira funções pequenas que podem ser verificadas com `assert`. Para projetos maiores, os mesmos testes podem migrar para `pytest`.

````python
def frequencia_por_palavra(texto: str) -> dict[str, int]:
    from collections import Counter
    return dict(Counter(texto.lower().split()))

assert frequencia_por_palavra("Python dados python") == {"python": 2, "dados": 1}
print("Teste da função de frequência: OK")
````
