````python
def simular(*respostas):
    """Devolve uma função que age como input(), mas lê de uma lista pronta."""
    it = iter(respostas)
    def entrada(msg=''):
        resp = next(it)
        print(f'{msg}{resp}')
        return resp
    return entrada
````

## 1. Listas

Uma lista guarda vários valores em ordem, entre colchetes `[ ]`. Aceita tipos misturados e **pode ser alterada** (é *mutável*).

````python
dados1 = [0, 2, 3, 'a']
dados2 = ['a', 'b', 0]
dados3 = dados1 + dados2     # + concatena listas
print(dados3)

print(len(dados3))           # quantidade de elementos
print(dados3.count('a'))     # quantas vezes 'a' aparece
print('a' in dados3)         # 'a' está na lista?
````

### Máximo, mínimo, soma e ordenação
`sorted(lista)` devolve **uma nova lista** ordenada e não mexe na original. Já `lista.sort()` ordena **a própria lista**.

````python
lista = [44, 8, -9, 23, -1, 66]
print(max(lista))
print(min(lista))
print(sum(lista))
print(sorted(lista))                  # nova lista
print(sorted(lista, reverse=True))    # ordem decrescente
print(lista)                          # a original continua igual
````

### Índices e fatiamento
Funcionam como nas strings: `lista[posição]`, `lista[início:fim:passo]`, e `-1` é o último.

````python
numeros = [10, 20, 30, 40, 50, 60]
print(numeros[0])       # primeiro
print(numeros[-1])      # último
print(numeros[1:4])     # posições 1, 2 e 3
print(numeros[::2])     # de 2 em 2
````

### `range` e listas
`range(início, fim, passo)` gera uma sequência de inteiros (sem incluir o `fim`). Para transformá-lo em lista, use `list(...)`.

⚠️ **Armadilha:** `[range(0, 5)]` cria uma lista com **um único elemento** (o objeto `range`), e não os números de 0 a 4.

````python
errado = [range(0, 5)]
print(errado)                  # [range(0, 5)]  <- não é o que queríamos

certo = list(range(0, 10))
print(certo)
````

### Métodos que alteram a lista

| Método | O que faz |
|---|---|
| `append(x)` | adiciona `x` no final |
| `insert(pos, x)` | insere `x` **antes** da posição `pos` |
| `remove(x)` | remove a **primeira** ocorrência do valor `x` |
| `pop(pos)` | remove e devolve o elemento da posição `pos` (sem argumento, o último) |
| `index(x)` | posição da primeira ocorrência de `x` |
| `count(x)` | quantas vezes `x` aparece |
| `sort()` / `reverse()` | ordena / inverte a própria lista |
| `clear()` | esvazia a lista |

````python
lista = [5, -2, 11, 3, 0, -2]

print('Adicionar o número 4 no final:')
lista.append(4)
print(lista)

print('Quantos números tem a lista?', len(lista))
print('Quantas vezes aparece o -2?', lista.count(-2))
print('Qual a posição do número 11?', lista.index(11))

print('Inserir o 12 na posição 2:')
lista.insert(2, 12)
print(lista)

print('Remover o -2 (só a primeira ocorrência):')
lista.remove(-2)
print(lista)

print('Remover o elemento da posição 3:', lista.pop(3))
print(lista)

print('Em ordem crescente:', sorted(lista))
````

### Listas dentro de listas
Cada elemento de uma lista pode ser outra lista. Para acessar, use dois índices: `lista[linha][posição]`.

Atenção a quem recebe o `append`: `lista.append(2)` coloca o 2 **na lista de fora**; `lista[0].append(2)` coloca na lista **de dentro**.

````python
lista = [[]]
lista.append(2)
print(lista)            # [[], 2]

lista[0].append(-1)
print(lista)            # [[-1], 2]

lista.append(5)
lista[0].append('a')
print(lista)            # [[-1, 'a'], 2, 5]
````

### ⚠️ Cuidado ao copiar listas
Em Python, `b = a` **não copia** a lista: as duas variáveis apontam para a *mesma* lista. Para copiar de verdade, use `a[:]` ou `a.copy()`.

Isso importa muito quando montamos uma "lista de listas" reaproveitando uma lista auxiliar:

````python
# Sem copiar: todos os elementos são a MESMA lista
aux = ['Ana', 30]
geral = []
geral.append(aux)
aux.clear()
print('Sem copiar:', geral)       # [[]]  <- perdemos os dados!

# Copiando com [:]
aux = ['Ana', 30]
geral = []
geral.append(aux[:])
aux.clear()
print('Copiando:  ', geral)       # [['Ana', 30]]
````

## 2. Tuplas

Parecidas com listas, mas **imutáveis**: depois de criadas não podem ser alteradas (não têm `append`, `remove`, etc.). Usam parênteses `( )`.

Use tupla para dados que **não devem mudar** (ex.: nomes dos meses, dias da semana).

````python
tupla = (4, -2, 8, -11, 8)

print(tupla[3])             # indexação igual à da lista
print(tupla.count(8))       # quantas vezes o 8 aparece
print(tupla.index(-2))      # posição do -2
print(len(tupla))
print(sorted(tupla))        # sorted() sempre devolve uma LISTA
````

````python
# Tentando alterar uma tupla:
try:
    tupla.append(7)
except AttributeError as erro:
    print('Erro:', erro)
````

## 3. Laço `for`

Repete um bloco **para cada elemento** de uma sequência. Use quando você já sabe quantas vezes (ou sobre o que) vai repetir.

O parâmetro `end=' '` do `print` troca a quebra de linha por um espaço, para imprimir tudo na mesma linha.

### Percorrendo uma lista

````python
lista = [1, 2, 3, 4]
for val in lista:
    print(val, end=' ')
````

### Percorrendo com `range`
- `range(5)` → 0, 1, 2, 3, 4
- `range(1, 5)` → 1, 2, 3, 4
- `range(0, 10, 2)` → 0, 2, 4, 6, 8
- `range(12, 0, -2)` → 12, 10, 8, 6, 4, 2 (passo negativo = contagem regressiva)

````python
for cont in range(5):
    print(cont, end=' ')
print()

for cont in range(1, 5):
    print(cont, end=' ')
print()

for cont in range(0, 10, 2):
    print(cont, end=' ')
print()

for cont in range(12, 0, -2):
    print(cont, end=' ')
````

````python
# Pares de 2 até um número dado
# num = int(input('Digite um número: '))
num = 15
for cont in range(2, num + 1, 2):      # num + 1 para o próprio num poder entrar
    print(cont, end=' ')
print()

# Início, fim e passo escolhidos pelo usuário
# i = int(input('Início: ')); f = int(input('Fim: ')); p = int(input('Passo: '))
i, f, p = 3, 20, 4
for cont in range(i, f, p):
    print(cont, end=' ')
````

### Índice e valor ao mesmo tempo
Três jeitos de percorrer uma lista. O `enumerate` é o mais prático quando precisamos da posição **e** do valor.

````python
lista = ['a', 8, 9, 'b']

print('1) direto nos valores:')
for valores in lista:
    print(valores, end=' ')

print('\n2) pelas posições, com range(len(...)):')
for posicao in range(len(lista)):
    print(f'posição {posicao} -> {lista[posicao]}')

print('3) com enumerate:')
for posicao, valores in enumerate(lista):
    print(f'posição {posicao} -> {valores}')
````

### `for` + `if`

### Padrão do acumulador e do contador
- **Contador:** começa em 0 e soma 1 a cada ocorrência.
- **Acumulador:** começa em 0 e soma o valor a cada volta.

**Ímpares entre 2 e 24 (com contagem)**

````python
quant = 0
for val in range(2, 25):
    if val % 2 != 0:
        quant += 1
        print(val, end=' ')
print(f'\nQuantidade de valores na faixa: {quant}')
````

````python
# Números entre 3 e 30 que deixam resto 2 na divisão por 3
qua = 0
for val in range(3, 31):
    if val % 3 == 2:
        qua += 1
        print(val, end=' ')
print(f'\nQuantidade de valores: {qua}')
````

````python
# Soma de 0 até valor-1  (o range NÃO inclui o valor final!)
# valor = int(input('Digite um valor: '))
valor = 5
soma = 0
for val in range(0, valor):
    soma += val
print(soma)          # 0+1+2+3+4 = 10

# Para somar até o próprio valor, use valor + 1
soma = 0
for val in range(0, valor + 1):
    soma += val
print(soma)          # 0+1+2+3+4+5 = 15
````

````python
# Ímpares múltiplos de 3 entre 0 e 99: quantidade e soma
qtde = 0
soma = 0
for val in range(0, 100):
    if val % 2 != 0 and val % 3 == 0:
        qtde += 1
        soma += val
        print(val, end=' ')
print(f'\nQuantidade: {qtde}')
print(f'Soma: {soma}')
````

### Maior e menor valor (sem usar `max`/`min`)
A ideia: assumir que o **primeiro** valor é o maior e o menor, e comparar com os demais.

````python
# Em vez de digitar 7 pesos, usamos uma lista pronta:
pesos = [72.5, 80.1, 65.3, 90.2, 55.0, 77.7, 68.9]

maior = menor = 0
for pos, peso in enumerate(pesos):
    if pos == 0:                # primeiro valor: ponto de partida
        maior = peso
        menor = peso
    else:
        if peso > maior:
            maior = peso
        elif peso < menor:
            menor = peso

print(f'Maior: {maior}')
print(f'Menor: {menor}')
print('Conferindo com max/min:', max(pesos), min(pesos))
````

## 4. Laço `while`

Repete **enquanto uma condição for verdadeira**. Use quando você **não sabe quantas vezes** vai repetir.

Todo `while` tem 3 partes:
```python
cont = 0              # 1. início
while cont < 5:       # 2. condição
    print(cont)
    cont = cont + 1   # 3. atualização (sem isso o laço nunca termina!)
```

| | `for` | `while` |
|---|---|---|
| Quando usar | sei quantas voltas / sobre o que percorrer | não sei quantas voltas |
| Controle | automático | você controla a condição |
| Risco | baixo | **laço infinito** se esquecer de atualizar |

Existem dois padrões muito comuns:
- **Condição de parada** no cabeçalho (`while cont <= num`).
- **`while True` + `break`**: roda "para sempre" até algo mandar sair.

````python
# Contagem de -5 a 3
cont = -5                    # início
while cont <= 3:             # condição
    print(cont, end=' ')
    cont = cont + 1          # atualização
print()

# De 8 até um número dado
# num = int(input('Digite um número: '))
num = 14
cont = 8
while cont <= num:
    print(cont, end=' ')
    cont = cont + 1
````

````python
# Soma dos pares de 0 até num
num = 10
cont = 0
soma = 0
while cont <= num:
    if cont % 2 == 0:
        soma += cont
        print(cont, end=' ')
    cont += 1
print(f'\nSoma dos valores pares: {soma}')
````

### `while True` + `break`: continuar [S/N]
Padrão clássico: lê um valor, pergunta se quer continuar e sai com `break` se a resposta for `N`.

````python
entrada = simular('2', 's', '3', 's', '5', 'n')      # <- apague para usar input()

soma = 0
while True:
    num = int(entrada('Digite o número: '))
    soma = soma + num
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print(f'Valor da soma: {soma}')
````

### Jogo de adivinhação (com dicas)
O computador "pensa" em um número e o jogador tenta até acertar, contando as tentativas.

````python
from random import randint

# compt = randint(1, 8)             # sorteio real
compt = 5                           # fixo, para o exemplo ser reproduzível
entrada = simular('2', '7', '5')    # palpites do jogador

cont = 1
while True:
    jogador = int(entrada('Digite um número: '))
    if jogador == compt:
        print('Acertou!')
        break
    else:
        print('Errou!', end=' ')
        if jogador > compt:
            print('Digite um número menor.')
        else:
            print('Digite um número maior.')
        cont += 1

print(f'Você precisou de {cont} tentativa(s).')
````

### Lista de números com estatísticas
Lemos números até o usuário parar, guardamos em uma lista e depois usamos as funções prontas.

````python
entrada = simular('4', 's', '7', 's', '10', 's', '1', 'n')

lista = []
while True:
    num = int(entrada('Digite um número: '))
    lista.append(num)
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print('Pares digitados:', [v for v in lista if v % 2 == 0])
print(f'Maior: {max(lista)}')
print(f'Menor: {min(lista)}')
print(f'Soma: {sum(lista)}')
print(f'Quantidade: {len(lista)}')
````

### Menu de opções
Um `while True` que só termina quando o usuário escolhe sair.

````python
# Respostas: números 3 e 4 -> soma -> multiplicação -> novos números (5 e 6) -> soma -> sair
entrada = simular('3', '4', '1', '2', '3', '5', '6', '1', '4')

num_1 = float(entrada('Digite o primeiro número: '))
num_2 = float(entrada('Digite o segundo número: '))

while True:
    print('\n[1] somar  [2] multiplicar  [3] novos números  [4] sair')
    per = int(entrada('Qual operação: '))
    if per == 1:
        print(f'Soma: {num_1 + num_2}')
    elif per == 2:
        print(f'Multiplicação: {num_1 * num_2}')
    elif per == 3:
        print('Informe novamente os números')
        num_1 = float(entrada('Digite o primeiro número: '))
        num_2 = float(entrada('Digite o segundo número: '))
    elif per == 4:
        print('Fim do programa')
        break
    else:
        print('Opção inválida')
````

### Tabuada (`for` dentro de `while`)

## 5. Listas na prática

### Lista sem repetidos, em ordem crescente e decrescente

````python
entrada = simular('5', 's', '3', 's', '5', 's', '9', 'n')

lista = []
while True:
    num = int(entrada('Digite o número: '))
    if num not in lista:
        lista.append(num)
        print('Número adicionado')
    else:
        print('Esse número já existe. Não adicionado')
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print(f'Números da lista: {lista}')
lista.sort()
print(f'Em ordem crescente: {lista}')
lista.sort(reverse=True)
print(f'Em ordem decrescente: {lista}')
````

### Inserção ordenada (sem usar `sort`)
A cada número novo, descobrimos **onde ele cabe** e usamos `insert`, mantendo a lista sempre ordenada.

````python
entrada = simular('7', '3', '9', '1', '5')

valores = []
for cont in range(5):
    num = int(entrada('Digite o número: '))
    if cont == 0:                       # primeiro número: só adiciona
        valores.append(num)
    elif num > valores[-1]:             # maior que o último: vai para o final
        valores.append(num)
    else:                               # senão, procura a posição certa
        pos = 0
        while pos < len(valores):
            if num <= valores[pos]:
                valores.insert(pos, num)     # insere ANTES dessa posição
                break
            pos += 1

print(valores)
````

### Validando parênteses em uma expressão
**Versão 1 — só contar.** Compara a quantidade de `(` com a de `)`. Funciona em muitos casos, mas falha em `)3+2(`, que tem a mesma quantidade mas está errada.

**Versão 2 — contador com verificação.** Soma 1 em cada `(` e subtrai 1 em cada `)`. Se o contador ficar **negativo** em algum momento, um `)` apareceu antes do `(` correspondente. No fim, o contador precisa ser **zero**.

````python
expressoes = ['(2+3)*(4-1)', '((2+3)', ')3+2(']

print('--- Versão 1: só contar ---')
for expressao in expressoes:
    a = expressao.count('(')
    b = expressao.count(')')
    print(f'{expressao:15} -> ', 'válida' if a == b else 'inválida')

print('\n--- Versão 2: contador com verificação ---')
for expressao in expressoes:
    cont = 0
    for caractere in expressao:
        if caractere == '(':
            cont += 1
        if caractere == ')':          # não usar elif: cada if é testado separadamente
            cont -= 1
        if cont < 0:                  # fechou sem ter aberto
            break
    print(f'{expressao:15} -> ', 'correta' if cont == 0 else 'incorreta')
````

### Cadastro de pessoas (lista de listas)
Cada pessoa é uma lista `[nome, idade]`; todas ficam dentro de uma lista geral. Note o uso de `[:]` para **copiar** a lista auxiliar antes de limpá-la.

````python
entrada = simular('Ana', '30', 's', 'Bruno', '25', 's', 'Carla', '30', 'n')

lista_geral = []
lista = []
while True:
    nome = str(entrada('Digite o nome: '))
    idade = int(entrada('Digite a idade: '))
    lista.append(nome)
    lista.append(idade)
    lista_geral.append(lista[:])       # cópia!
    lista.clear()
    per = str(entrada('Continuar [S/N]: ')).strip().upper()[0]
    if per == 'N':
        break

print('-' * 50)
print(lista_geral)
print(f'Quantidade de pessoas: {len(lista_geral)}')

# Maior idade (primeira pessoa como ponto de partida)
maior = lista_geral[0][1]
for pessoa in lista_geral:
    if pessoa[1] > maior:
        maior = pessoa[1]

print(f'Maior idade: {maior}, de:', end=' ')
for pessoa in lista_geral:
    if pessoa[1] == maior:
        print(pessoa[0], end='  ')
````

### List comprehension
Forma compacta de criar listas: `[expressão for item in sequência if condição]`.

````python
print([x * 2 for x in [1, 2, 3, 4]])                  # dobro de cada um
print([(x, x * 2) for x in [1, 2, 3, 4]])             # lista de tuplas
print([x for x in range(1, 21) if x % 3 == 0])        # só os múltiplos de 3
````

### Revisão rápida: o dobro dos dados

````python
entrada = simular('4', '10', '7')

dados = []
dados_dobro = []
for pos in range(3):
    num = int(entrada(f'Na posição {pos}, digite um número: '))
    dados.append(num)
    dados_dobro.append(2 * num)

print(f'Dados: {dados}')
print(f'Dobro: {dados_dobro}')
````

## Aprofundamento 1 — Complexidade, coleções e legibilidade
Em listas, procurar um elemento geralmente percorre até `n` posições: custo `O(n)`. Em conjuntos (`set`), a verificação de pertinência costuma ter custo médio `O(1)`, embora isso não seja uma garantia universal.

````python
nomes = ["Ana", "Bruno", "Ana", "Carla", "Bruno"]
unicos_na_ordem = list(dict.fromkeys(nomes))
consulta_rapida = set(nomes)
assert unicos_na_ordem == ["Ana", "Bruno", "Carla"]
assert "Bruno" in consulta_rapida
print("Ordem preservada:", unicos_na_ordem)
````

## Aprofundamento 2 — Compreensões e geradores
Compreensões criam coleções; expressões geradoras produzem valores sob demanda e podem poupar memória.

````python
quadrados_pares = [n*n for n in range(1, 11) if n % 2 == 0]
soma_quadrados = sum(n*n for n in range(1, 11))
assert quadrados_pares == [4, 16, 36, 64, 100]
assert soma_quadrados == 385
print(quadrados_pares, soma_quadrados)
````

## Exercício adicional (com solução) — contagem por categoria
Dada uma lista de cidades, gere uma contagem sem perder a distinção entre nomes e apresente o resultado ordenado por quantidade.

````python
from collections import Counter
cidades = ["Recife", "Campinas", "Recife", "Salvador", "Campinas", "Recife"]
ranking = Counter(cidades).most_common()
assert ranking[0] == ("Recife", 3)
print(ranking)
````
