# Widgets e interatividade no JupyterLab

**Trilha complementar | Nível:** introdutório → intermediário.

**Objetivos:** criar controles reativos, compreender callbacks, tratar uploads e estruturar interfaces exploratórias seguras. Os 19 exemplos e explicações originais foram preservados e ampliados.

```python
try:
    import ipywidgets as widgets
    from IPython.display import display, clear_output
    DISPONIVEL = True
except ImportError:
    DISPONIVEL = False
    print('Instale ipywidgets no ambiente do kernel para executar os exemplos.')
```

## 1. Valores numéricos, intervalos e progresso

Sliders controlam valores individuais ou intervalos. Textos numéricos podem ter limites. Barras de progresso exibem valores de um processo.

```python
if DISPONIVEL:
    display(widgets.IntSlider(value=5, min=0, max=10, step=1, description='Inteiro'))
    display(widgets.FloatSlider(value=0.5, min=0, max=1, step=0.05, description='Decimal'))
    display(widgets.IntRangeSlider(value=(2, 7), min=0, max=10, description='Intervalo'))
    display(widgets.IntProgress(value=4, min=0, max=10, description='Progresso'))
    display(widgets.BoundedIntText(value=5, min=0, max=10, description='Limitado'))
```

## 2. Escolhas, booleanos e entradas de texto

`value` contém o valor atual, `description` define rótulo e `disabled` impede interação. A escolha do componente depende do tipo de valor e da quantidade de opções.

```python
if DISPONIVEL:
    display(widgets.Checkbox(value=True, description='Ativar recurso', indent=False))
    display(widgets.ToggleButton(value=False, description='Alternar'))
    display(widgets.Dropdown(options=['Pandas', 'NumPy', 'Matplotlib'], value='Pandas', description='Biblioteca'))
    display(widgets.RadioButtons(options=['Baixo', 'Médio', 'Alto'], description='Prioridade'))
    display(widgets.SelectMultiple(options=['Python', 'SQL', 'Git'], description='Interesses'))
    display(widgets.Text(value='Olá!', description='Texto'))
    display(widgets.Textarea(value='Anotações...', description='Notas'))
    display(widgets.Combobox(placeholder='Digite ou escolha', options=['Python', 'Pandas'], description='Busca'))
```

## 3. Componentes adicionais

`DatePicker` seleciona datas; `FileUpload` disponibiliza arquivos enviados ao kernel — **não imprima conteúdos ou senhas em notebooks públicos**. Os formatos de `FileUpload.value` podem variar entre versões. Componentes como `Image` aceitam bytes de imagem; aqui geramos um SVG pequeno localmente, sem depender de arquivos externos.

```python
if DISPONIVEL:
    display(widgets.DatePicker(description='Data'))
    display(widgets.FileUpload(accept='.csv', multiple=False, description='Enviar CSV'))
    svg = b'<svg xmlns="http://www.w3.org/2000/svg" width="160" height="50"><rect width="160" height="50" fill="#dfebf5"/><text x="12" y="30" font-size="16">Exemplo SVG</text></svg>'
    display(widgets.Image(value=svg, format='svg+xml', width=160, height=50))
```

## 4. Layout: `HBox`, `VBox`, `GridBox`, `Accordion` e `Tab`

A composição de widgets organiza os controles em linhas, colunas e abas. Use layouts para separar filtros de resultados.

```python
if DISPONIVEL:
    a = widgets.IntSlider(description='A')
    b = widgets.FloatSlider(description='B')
    display(widgets.HBox([a, b]))
    display(widgets.VBox([widgets.Label('Filtros'), widgets.Checkbox(description='Incluir nulos'), widgets.Text(description='Nome')]))
    grade = widgets.GridBox([widgets.Button(description=str(i)) for i in range(1, 5)], layout=widgets.Layout(grid_template_columns='repeat(2, 110px)', grid_gap='6px'))
    display(grade)
    abas = widgets.Tab(children=[widgets.Label('Aba um'), widgets.Label('Aba dois')])
    abas.set_title(0, 'Dados'); abas.set_title(1, 'Resultado')
    display(abas)
    acord = widgets.Accordion(children=[widgets.Label('Configuração avançada')])
    acord.set_title(0, 'Opções')
    display(acord)
```

## 5. Interatividade com `interact` e eventos

`interact` relaciona widgets a parâmetros de funções. `observe` reage a mudanças de valor; `on_click` conecta botões a callbacks. Use `Output` para controlar onde os resultados aparecem.

```python
if DISPONIVEL:
    @widgets.interact(fator=widgets.IntSlider(value=2, min=1, max=10))
    def tabela_simples(fator):
        print([fator * i for i in range(1, 6)])

    botao = widgets.Button(description='Calcular dobro')
    entrada = widgets.IntText(value=5, description='Valor')
    saida = widgets.Output()
    def clicar(_):
        with saida:
            clear_output(wait=True)
            print('Dobro:', entrada.value * 2)
    botao.on_click(clicar)
    display(widgets.VBox([entrada, botao, saida]))
```

## 6. Visualização controlada por sliders

Um slider pode controlar uma curva Matplotlib. A célula só abre a visualização quando executada no JupyterLab; em visualizadores estáticos, o código permanece disponível para estudo.

```python
if DISPONIVEL:
    try:
        import numpy as np
        import matplotlib.pyplot as plt
    except ImportError:
        print('Instale numpy e matplotlib para executar o gráfico interativo.')
    else:
        @widgets.interact(frequencia=widgets.FloatSlider(min=0.5, max=5, step=0.5, value=1, description='Frequência'))
        def onda(frequencia):
            x = np.linspace(0, 2 * np.pi, 300)
            fig, ax = plt.subplots(figsize=(6, 2.5))
            ax.plot(x, np.sin(frequencia * x))
            ax.set_title('Seno com frequência variável')
            ax.grid(alpha=0.3)
            plt.show()
```

## 7. Exercícios resolvidos

**Exercício 1 — Conversor de temperatura:** crie um slider de Celsius e atualize automaticamente a leitura em Fahrenheit.

```python
def celsius_para_fahrenheit(c):
    return c * 9 / 5 + 32

assert celsius_para_fahrenheit(0) == 32
assert celsius_para_fahrenheit(100) == 212

if DISPONIVEL:
    celsius = widgets.FloatSlider(min=-20, max=100, value=25, description='°C')
    resultado = widgets.HTML()
    def atualizar(change=None):
        resultado.value = f'<b>{celsius.value:.1f} °C = {celsius_para_fahrenheit(celsius.value):.1f} °F</b>'
    celsius.observe(atualizar, names='value')
    atualizar()
    display(widgets.VBox([celsius, resultado]))
```

**Exercício 2 — Filtrar itens:** selecione uma categoria para ver quais registros correspondem a ela, sem depender de dados externos.

```python
registros = [
    {'nome': 'Caderno', 'categoria': 'Papelaria'},
    {'nome': 'Caneta', 'categoria': 'Papelaria'},
    {'nome': 'Mouse', 'categoria': 'Informática'},
]
def filtrar_categoria(categoria):
    return [r['nome'] for r in registros if r['categoria'] == categoria]

assert filtrar_categoria('Informática') == ['Mouse']
if DISPONIVEL:
    seletor = widgets.Dropdown(options=['Papelaria', 'Informática'], description='Categoria')
    lista = widgets.HTML()
    def mostrar(change=None):
        lista.value = '<ul>' + ''.join(f'<li>{nome}</li>' for nome in filtrar_categoria(seletor.value)) + '</ul>'
    seletor.observe(mostrar, names='value')
    mostrar()
    display(widgets.VBox([seletor, lista]))
```

## 8. Modelo de eventos: `observe`, `on_click` e `Output`

Widgets mantêm **estado** e notificam alterações. A função passada a `.observe()` recebe um dicionário `change`, contendo, entre outros campos, `name`, `old` e `new`. Use `names='value'` para evitar reagir a alterações não relevantes. `Button.on_click` recebe o botão clicado, enquanto `Output` permite agrupar a saída em uma região específica.

**Boas práticas:** evite registrar o mesmo callback várias vezes ao executar a célula repetidamente; use `.unobserve` ou `.on_click(..., remove=True)` ao desmontar interfaces. Feche widgets descartados quando apropriado; resultados dependem do frontend Jupyter.

```python
# Exemplo isolado: não precisa de rede nem de arquivos.
if DISPONIVEL:
    contador = widgets.IntSlider(value=0, min=0, max=10, description='Contador')
    painel = widgets.Output()
    def apresentar_contador(change):
        with painel:
            clear_output(wait=True)
            print(f'Valor atual: {change["new"]}')
    contador.observe(apresentar_contador, names='value')
    display(widgets.VBox([contador, painel]))
    apresentar_contador({'new': contador.value})
else:
    print('Instale ipywidgets no kernel para o exemplo visual.')
```

## 10. Exercício integrador: filtro de uma pequena tabela

**Objetivo:** selecionar uma categoria e limitar um valor numérico, combinando `Dropdown`, `IntSlider` e `Output`. O conjunto é fictício e gerado no próprio notebook. Não há dependência de Pandas.

```python
itens = [
    {'produto': 'Caderno', 'categoria': 'Papelaria', 'preco': 18},
    {'produto': 'Caneta', 'categoria': 'Papelaria', 'preco': 5},
    {'produto': 'Mouse', 'categoria': 'Informática', 'preco': 65},
    {'produto': 'Teclado', 'categoria': 'Informática', 'preco': 90},
]
def filtrar_itens(categoria, preco_max):
    return [i for i in itens if (categoria == 'Todas' or i['categoria'] == categoria) and i['preco'] <= preco_max]
assert len(filtrar_itens('Papelaria', 20)) == 2
assert len(filtrar_itens('Informática', 70)) == 1
if DISPONIVEL:
    categoria = widgets.Dropdown(options=['Todas','Papelaria','Informática'],description='Categoria')
    preco = widgets.IntSlider(value=100,min=0,max=100,description='Até R$')
    resultado = widgets.Output()
    def atualizar(_=None):
        with resultado:
            clear_output(wait=True)
            for item in filtrar_itens(categoria.value,preco.value):
                print(f"{item['produto']}: R$ {item['preco']}")
    categoria.observe(atualizar,names='value')
    preco.observe(atualizar,names='value')
    display(widgets.VBox([categoria,preco,resultado]))
    atualizar()
else:
    print(filtrar_itens('Todas', 50))
```

## 11. Limitações e boas práticas

- Interatividade depende do frontend e de uma versão compatível de `ipywidgets`; a existência do widget no código **não garante** renderização em todos os ambientes.
- Evite tarefas pesadas a cada mudança de slider: use cache, amostragem ou um botão para confirmar a atualização.
- Limpe saídas com parcimônia para não perder mensagens importantes de erro.
- Não grave arquivos automaticamente apenas porque um controle mudou.
- Para leitores estáticos (GitHub ou PDF), inclua uma descrição textual e um resultado não interativo.
