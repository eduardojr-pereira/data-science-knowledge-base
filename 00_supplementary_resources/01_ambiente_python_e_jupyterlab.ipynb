## 5. Como o JupyterLab encontra o Python

O **servidor JupyterLab** oferece a interface; o **kernel** executa o código. Eles podem usar Python diferentes. Um pacote instalado no terminal não aparece necessariamente no kernel. `sys.executable` identifica o interpretador *do kernel*, e `sys.prefix` indica o ambiente ativo.

Em uma célula Python, prefira `%pip install pacote` quando precisar instalar algo: o magic usa o interpretador do kernel. **Não execute instalações automaticamente neste guia.**

Para cadastrar um ambiente como kernel, no terminal com o ambiente ativado: `python -m ipykernel install --user --name meu-ambiente --display-name "Python (meu-ambiente)"`. Isso cria um registro de kernel (ação de configuração, não necessária para seguir os exemplos).

```python
import sys
from pathlib import Path
print('Kernel executa:', Path(sys.executable))
print('Ambiente:', Path(sys.prefix))
print('Ambiente virtual?', sys.prefix != sys.base_prefix)
assert Path(sys.executable).exists()
```

## 6. Dependências, versões e reprodução

- `python -m pip list` mostra pacotes instalados no interpretador escolhido.
- `python -m pip check` informa conflitos declarados entre dependências; **não** garante que o notebook funciona.
- `python -m pip freeze > ambiente_completo.txt` captura o ambiente inteiro: útil para auditoria, mas geralmente **não** deve substituir um `requirements.txt` pequeno e específico do projeto.
- `python -m pip install -r requirements.txt` instala os requisitos declarados, preferencialmente em um ambiente novo de testes.

Evite atualizações globais, especialmente em ambiente compartilhado por diferentes disciplinas.

```python
from importlib import metadata
pacotes = ['ipykernel', 'jupyterlab', 'numpy', 'pandas', 'polars', 'ipywidgets']
for pacote in pacotes:
    try:
        print(f'{pacote:12} {metadata.version(pacote)}')
    except metadata.PackageNotFoundError:
        print(f'{pacote:12} não instalado neste kernel')
```

## 7. Diretórios, caminhos e preservação de dados

O diretório atual (`Path.cwd()`) define a resolução de caminhos relativos. Ele pode mudar conforme a forma de abrir o JupyterLab. Prefira `Path('datasets') / 'arquivo.csv'` a caminhos absolutos da sua máquina. Não altere arquivos de origem: escreva resultados em `outputs/` e faça cópias em locais apropriados.

```python
from pathlib import Path
base = Path.cwd()
print('Diretório de execução:', base)
print('Pasta datasets existe?', (base/'datasets').is_dir())
print('Pasta outputs existe?', (base/'outputs').is_dir())
# Apenas inspeciona; não cria, modifica ou remove arquivos.
```

## 8. Exercício de diagnóstico (resolvido)

**Problema:** um notebook importa um pacote pelo terminal, mas falha ao importá-lo no JupyterLab. Que evidências coletar **antes** de reinstalar?

**Solução comentada:** identificar `sys.executable` e `sys.prefix`; comparar com `which python` e `python -m pip show pacote` no terminal; conferir o kernel escolhido; reiniciar o kernel depois de instalar; usar `%pip` apenas quando necessário.

**Limitação:** um pacote pode estar instalado e ainda falhar por dependências nativas, incompatibilidade de ABI ou configuração do sistema.

```python
import sys
from pathlib import Path
def resumo_ambiente():
    return {'python': sys.version.split()[0], 'executavel': str(Path(sys.executable)), 'diretorio': str(Path.cwd())}
diagnostico = resumo_ambiente()
assert Path(diagnostico['executavel']).is_file()
assert Path(diagnostico['diretorio']).is_dir()
print(diagnostico)
```
