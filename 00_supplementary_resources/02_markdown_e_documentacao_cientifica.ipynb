## Títulos e ênfase

Em células Markdown, use `#` para títulos, `**texto**` para negrito e `*texto*` para itálico.

````python
exemplo = "# Título\n\n**Negrito**, *itálico* e `código`."
print(exemplo)
````

## Listas, citações e tabelas

Listas começam com `-` ou números; citações com `>`; tabelas usam barras verticais.

````python
exemplo = "| Coluna | Valor |\n|---|---|\n| A | 1 |"
print(exemplo)
````

## Código e matemática

Blocos de código usam três crases. Fórmulas podem usar sintaxe LaTeX no Markdown.

````python
exemplo = "```python\nprint('Olá')\n```"
print(exemplo)
````

## Exercício resolvido — Documente uma análise

**Enunciado:** Crie um pequeno relatório Markdown a partir de dados.

**Solução completa:**

````python
resumo = {"titulo":"Análise exploratória", "objetivo":"Descrever os dados"}
texto = f"# {resumo['titulo']}\n\n**Objetivo:** {resumo['objetivo']}"
print(texto)
````

## Matemática em Markdown com LaTeX

Uma expressão inline, como $\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i$, é útil no texto. Equações em bloco recebem destaque:

$$
\mathrm{RMSE}=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}
$$

O RMSE penaliza erros grandes mais intensamente do que a média absoluta dos erros. **Defina os símbolos:** $n$ é o número de exemplos, $y_i$ é o valor observado e $\hat y_i$ é a previsão.\n\nCuidado para não confundir Markdown, LaTeX e código Python: cada um tem sua sintaxe e finalidade.

## Relatório científico — modelo comentado

### Pergunta de pesquisa

Qual é a distribuição das observações e quais limitações afetam sua interpretação?

### Método

Descrever fonte, período, tamanho da amostra, critérios de exclusão e procedimentos de tratamento. Código e gráficos devem ter títulos, unidades e origem dos dados.

### Resultados

Relatar observações mensuráveis e a incerteza quando pertinente. Tabelas não substituem interpretação.

### Limitações e conclusão

Registrar vieses, amostragem, valores faltantes e condições sob as quais a análise não se generaliza.

````python
# Exemplo independente: gerar o esqueleto Markdown de um relatório em memória.
from textwrap import dedent
relatorio = dedent("""# Relatório exploratório

## Objetivo
Descrever uma amostra ilustrativa.

## Métodos
Conferir qualidade dos dados e produzir estatísticas descritivas.

## Resultados
Inserir tabela e interpretação com unidades.

## Limitações
A amostra pode não representar a população.
""")
assert relatorio.count('## ') == 4
print(relatorio)
````
