## Demonstração 2 — validação e limitações

```python
# Laboratório Spark real opcional, executado em interpretador isolado.
# Ative SOMENTE quando o ambiente PySpark e o Java já estiverem preparados.
import os, sys, subprocess, textwrap
if os.environ.get('RUN_SPARK_LABS') == '1':
    interpretador = os.environ.get('SPARK_LAB_PYTHON', sys.executable)
    assert os.path.isfile(interpretador), f'Interpretador não encontrado: {interpretador}'
    programa = textwrap.dedent(r'''from pyspark.sql import SparkSession, functions as F
from pyspark.sql.types import StructType,StructField,LongType,DoubleType,StringType
from pathlib import Path
s=SparkSession.builder.master('local[2]').appName('ep-area10-08').config('spark.ui.enabled','false').getOrCreate()
try:
    s.sparkContext.setLogLevel('ERROR')
    campos='year month day dep_time dep_delay arr_time arr_delay carrier tailnum flight origin dest air_time distance hour minute'.split()
    texto={'carrier','tailnum','origin','dest'}
    inteiro={'year','month','day','flight','hour','minute'}
    esquema=StructType([StructField(c,StringType() if c in texto else LongType() if c in inteiro else DoubleType(),True) for c in campos])
    df=s.read.option('header','true').schema(esquema).csv(str(Path('datasets/flights.csv').resolve()))
    n=df.filter(F.col('distance')>1000).count()
    import pandas as pd
    ref=int((pd.to_numeric(pd.read_csv('datasets/flights.csv')['distance'],errors='coerce')>1000).sum())
    assert n==ref,(n,ref)
    assert df.schema['distance'].dataType.typeName()=='double'
    print('Esquema explícito, voos >1000:',n,'| referência pandas:',ref)
    print('SPARK_LAB_OK_08')
finally:s.stop()''')
    ambiente = os.environ.copy()
    ambiente.setdefault('SPARK_LOCAL_IP', '127.0.0.1')
    ambiente['PYSPARK_PYTHON'] = interpretador
    ambiente['PYSPARK_DRIVER_PYTHON'] = interpretador
    resultado = subprocess.run([interpretador, '-c', programa], cwd=os.getcwd(),
                               env=ambiente, text=True, capture_output=True, timeout=150)
    if resultado.stdout: print(resultado.stdout.strip())
    if resultado.returncode:
        print(resultado.stderr[-3500:])
        raise RuntimeError('O LABORATÓRIO SPARK 08 FALHOU; não registrar como aprovado.')
    assert 'SPARK_LAB_OK_08' in resultado.stdout, 'Marcador de sucesso Spark não encontrado'
else:
    print('SPARK NÃO EXECUTADO (opcional). Ative RUN_SPARK_LABS=1 e SPARK_LAB_PYTHON.')
```
