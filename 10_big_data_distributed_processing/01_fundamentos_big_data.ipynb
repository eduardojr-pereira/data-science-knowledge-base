## Demonstração 1 — construção e observação

```python
from pathlib import Path
import pandas as pd
p = Path('datasets/flights.csv')
voos = pd.read_csv(p)
assert {'carrier','origin','dest','dep_delay','arr_delay','air_time'}.issubset(voos.columns)
print('Linhas:', len(voos), 'Colunas:', len(voos.columns))
print('Memória aproximada (MB):', round(voos.memory_usage(deep=True).sum()/1e6, 3))
print(voos[['carrier','origin','dest']].head())
```

## Interpretação

## Demonstração 2 — validação e limitações

```python
print('Dados ausentes, por coluna:')
print(voos[['dep_delay','arr_delay','air_time']].isna().mean().round(3))
print('Cuidado: este CSV pequeno serve para demonstrar conceitos, não simula um cluster real.')
```

## Boas práticas e limitações

- Execute com versões registradas, datasets identificados e caminhos relativos.
