## Demonstração 1 — construção e observação

```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus=['dados distribuídos em cluster','spark e processamento distribuído',
        'hadoop e armazenamento de dados']
vet=TfidfVectorizer(norm=None, smooth_idf=True)
X=vet.fit_transform(corpus)
print(vet.get_feature_names_out())
print(X.toarray().round(3))
```

## Interpretação
Observe o esquema, as unidades e os resultados. Antes de concluir que um algoritmo é eficiente, distinga **complexidade computacional**, **custo de comunicação** e **volume de dados**. Os cálculos locais ilustram a semântica; o Spark, quando disponível, ilustra a API distribuída.

## Demonstração 2 — validação e limitações

```python
import numpy as np
n = len(corpus)
termo = 'dados'
df = sum(termo in doc.lower().split() for doc in corpus)
idf = np.log((1+n)/(1+df)) + 1
indice = list(vet.get_feature_names_out()).index(termo)
assert np.isclose(vet.idf_[indice],idf), (vet.idf_[indice],idf)
print('DF:',df,'| IDF suavizado manual:',round(idf,3),'| scikit-learn:',round(vet.idf_[indice],3))
print('TF-IDF local: não houve MapReduce distribuído neste notebook.')
```
