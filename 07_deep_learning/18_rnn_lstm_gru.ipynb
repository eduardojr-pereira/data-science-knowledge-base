RNNs processam sequências com estado oculto compartilhado. O gradiente pode desaparecer/explodir em sequências longas; LSTM e GRU usam portas para reter e atualizar informação. Shapes e máscaras distinguem sequências de comprimentos variáveis.

## 2. Matemática essencial

**Expressão esquemática:** `h_t=tanh(W_x x_t+W_h h_(t−1)+b)`

A expressão acima resume a relação central. Para a derivação, identifique dimensões, hipóteses e operações; não interprete notação compacta como uma implementação completa.

## 3. Experimento obrigatório — NumPy e/ou scikit-learn

Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
import numpy as np
rng=np.random.default_rng(0)
x=np.array([.2,.3,.4]);h=0.;wx=.8;wh=.35
estados=[]
for t in x:
 h=np.tanh(wx*t+wh*h);estados.append(h)
print('Estados da RNN escalar:',np.round(estados,4).tolist())
assert len(estados)==3
```

## Laboratório aplicado revisado — Sensibilidade a estados iniciais em RNN escalar

Estados recorrentes, gradientes no tempo e portas de memória ajudam a explicar RNN/LSTM/GRU e suas limitações.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
seq=np.ones(25)*.1
def rnn(h0,wh):
 h=h0
 for x in seq:h=np.tanh(.2*x+wh*h)
 return h
h=1e-5
for wh in [.2,.6,.95]:
 derivative=(rnn(h,wh)-rnn(-h,wh))/(2*h)
 print('peso recorrente',wh,'sensibilidade ao estado inicial',round(derivative,8))
assert abs(rnn(h,.2)-rnn(-h,.2))<.1
```

### Interpretação, limites e verificação

É um experimento ilustrativo, não diagnóstico universal de gradientes: saturação de tanh e pesos recorrentes podem alterar o comportamento.
