## 3. Experimento obrigatório — NumPy e/ou scikit-learn

Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
import numpy as np
rng=np.random.default_rng(5)
x=rng.normal(size=(3,4));W=rng.normal(size=(4,4))*.1
res=x+np.tanh(x@W)
assert res.shape==x.shape
print('Bloco residual (demonstração de shapes):',x.shape,'→',res.shape)
```

## Laboratório aplicado revisado — Extração de características e distinção de transferência real

Transfer learning exige pesos pré-treinados e alinhamento de pré-processamento; congelar pesos aleatórios não é transferência.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
rng=np.random.default_rng(16)
features=rng.normal(size=(25,6))
fixed=rng.normal(size=(6,4))
representation=np.maximum(features@fixed,0)
print('Representação fixa (demonstração):',representation.shape)
print('ATENÇÃO: matriz aleatória não é backbone pré-treinado; não constitui transfer learning.')
assert representation.shape==(25,4)
```

### Interpretação, limites e verificação

Transferência requer pesos aprendidos previamente em outra tarefa. Um backbone inicializado aleatoriamente e congelado NÃO transfere conhecimento; documentar fonte, preprocessamento e licença dos pesos.
