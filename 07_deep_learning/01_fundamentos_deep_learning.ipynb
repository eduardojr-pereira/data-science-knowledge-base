### Estudo dirigido — profundidade, parâmetros e custo

Um modelo profundo combina sucessivas transformações para construir representações. **Profundidade por si só não garante generalização:** capacidade, dados, regularização, otimização e protocolo de avaliação são igualmente importantes. Nesta atividade, comparamos arquiteturas por quantidade de parâmetros, sem alegar que uma delas é superior em dados reais.

Para uma camada densa com `n_entrada` variáveis e `n_saida` neurônios, contamos `n_entrada × n_saida + n_saida` parâmetros (pesos e vieses).

```python
from math import prod

def parametros_densas(larguras):
    """Parâmetros treináveis de camadas Dense consecutivas (com bias)."""
    if len(larguras) < 2 or any(n <= 0 for n in larguras):
        raise ValueError("Forneça pelo menos duas larguras inteiras positivas.")
    return sum(a * b + b for a, b in zip(larguras[:-1], larguras[1:]))

modelos = {"raso": [8, 16, 1], "mais_profundo": [8, 8, 8, 1], "mais_largo": [8, 32, 1]}
for nome, camadas in modelos.items():
    print(f"{nome:15} camadas={camadas} parâmetros={parametros_densas(camadas)}")
assert parametros_densas([8, 16, 1]) == (8*16+16)+(16*1+1)
assert parametros_densas([8, 8, 8, 1]) == (8*8+8)+(8*8+8)+(8*1+1)
```

#### Análise e exercícios

**Interpretação:** conte parâmetros e compare largura/profundidade. Um modelo com menos parâmetros pode funcionar melhor, pior ou de modo semelhante; somente experimentos controlados com dados e avaliação separados poderiam sustentar essa comparação. **Atividade:** proponha como testaria três arquiteturas mantendo a mesma divisão de dados, pré-processamento, orçamento e métricas.
