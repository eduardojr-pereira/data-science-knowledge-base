### Experimento completo — pipeline, baseline e avaliação sem vazamento

Este experimento ilustra a ordem correta das operações: **divisão antes de ajuste**, transformação aprendida apenas no treino, baseline e avaliação em dados não vistos. O dataset Breast Cancer está integrado ao scikit-learn, sem download. O experimento usa regressão logística como referência de preparação de dados; **não representa treinamento de rede neural**.

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, confusion_matrix

X, y = load_breast_cancer(return_X_y=True)
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
modelo = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, random_state=42))
modelo.fit(X_treino, y_treino)
previsoes = modelo.predict(X_teste)
baseline = DummyClassifier(strategy='most_frequent').fit(X_treino, y_treino)
prev_baseline = baseline.predict(X_teste)

scaler_treinado = modelo.named_steps['standardscaler']
assert np.allclose(scaler_treinado.mean_, X_treino.mean(axis=0))
assert len(y_treino) + len(y_teste) == len(y)
print('Acurácia baseline:', round(accuracy_score(y_teste, prev_baseline), 3))
print('Acurácia pipeline:', round(accuracy_score(y_teste, previsoes), 3))
print('Acurácia balanceada:', round(balanced_accuracy_score(y_teste, previsoes), 3))
print('Matriz de confusão (linhas=classe real):')
print(confusion_matrix(y_teste, previsoes))
```

#### Análise e exercícios

**Interpretação:** o scaler está dentro do pipeline e aprende estatísticas só de `X_treino`. O teste é usado apenas para avaliação. O conjunto é didático; acurácia e acurácia balanceada não substituem avaliação externa, análise de risco ou validação clínica. **Atividade:** use uma validação cruzada estratificada *somente no treino* para comparar duas configurações e preserve `X_teste` até o final.
