```python
import numpy as np
X=np.arange(12,dtype='float32').reshape(4,3)
W=np.array([[1.,0.],[0.,1.],[-1.,2.]],dtype='float32')
Z=X@W
assert Z.shape==(4,2)
print('X',X.shape,'W',W.shape,'XW',Z.shape,'dtype',Z.dtype)
```

### Laboratório reproduzível — broadcasting, dtype e formas

Multiplicação elemento a elemento, produto matricial e *broadcasting* têm semânticas distintas. A ordem dos eixos importa para imagens: no formato `channels_last`, um lote RGB possui forma `(B, H, W, 3)`. Os exemplos seguintes usam apenas NumPy, garantindo execução sem TensorFlow.

```python
import numpy as np

lote = np.arange(2 * 3 * 4 * 3, dtype=np.float32).reshape(2, 3, 4, 3)
media_por_canal = lote.mean(axis=(0, 1, 2))
centrado = lote - media_por_canal  # broadcasting: (3,) -> (2,3,4,3)
assert centrado.shape == (2, 3, 4, 3)
assert np.allclose(centrado.mean(axis=(0, 1, 2)), 0, atol=1e-6)

X = np.arange(12, dtype=np.float32).reshape(4, 3)
W = np.arange(6, dtype=np.float32).reshape(3, 2)
Z = X @ W
assert Z.shape == (4, 2)
print('lote:', lote.shape, 'média por canal:', media_por_canal)
print('produto matricial:', Z.shape, 'dtype:', Z.dtype)

try:
    _ = X @ np.ones((4, 2), dtype=np.float32)
except ValueError as erro:
    print('Shape incompatível detectado corretamente:', type(erro).__name__)
else:
    raise AssertionError('Era esperado erro por dimensões incompatíveis.')
```

#### Análise e exercícios

**Interpretação:** eixo do lote (`axis=0`) não é eixo de canal (`axis=-1`). O NumPy informa incompatibilidade de forma; não se deve silenciar isso com um `reshape` arbitrário. **Atividade:** explique como as formas mudariam em `channels_first` e quando seria necessário transpor os eixos.
