### Laboratório — limites do perceptron em AND e XOR

O perceptron de uma camada pode aprender funções linearmente separáveis, mas não resolver XOR corretamente em todas as quatro combinações. Para tornar isso observável, treinamos a mesma implementação com funções AND e XOR, sem alterar o algoritmo entre os testes.

```python
import numpy as np

X = np.array([[0., 0.], [0., 1.], [1., 0.], [1., 1.]])
rotulos = {'AND': np.array([0, 0, 0, 1]), 'XOR': np.array([0, 1, 1, 0])}

def treinar_perceptron(X, y, taxa=0.2, epocas=100):
    pesos = np.zeros(X.shape[1], dtype=float)
    vies = 0.0
    historico = []
    for _ in range(epocas):
        erros = 0
        for xi, yi in zip(X, y):
            previsto = int(xi @ pesos + vies >= 0)
            delta = int(yi) - previsto
            pesos += taxa * delta * xi
            vies += taxa * delta
            erros += int(delta != 0)
        historico.append(erros)
    return pesos, vies, historico

for nome, y in rotulos.items():
    w, b, historico = treinar_perceptron(X, y)
    previsto = (X @ w + b >= 0).astype(int)
    acertos = int(np.sum(previsto == y))
    print(nome, 'previsões:', previsto.tolist(), 'acertos:', f'{acertos}/4',
          'erros na última época:', historico[-1])
    if nome == 'AND':
        assert np.array_equal(previsto, y)
    else:
        assert acertos < 4, 'Um único perceptron não deve resolver XOR.' 
```

#### Análise e exercícios

**Interpretação:** a contagem de erros ao final do treino é um diagnóstico do conjunto didático, não uma métrica de generalização. XOR requer representação não linear, como uma rede com camada oculta e ativação apropriada. **Atividade:** desenhe as quatro observações no plano e justifique por que uma única reta não separa XOR.
