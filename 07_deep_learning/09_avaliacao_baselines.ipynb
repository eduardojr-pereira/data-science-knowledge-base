## 3. Experimento obrigatório — NumPy e/ou scikit-learn

Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
X,y=load_breast_cancer(return_X_y=True)
a,b,ya,yb=train_test_split(X,y,random_state=9,stratify=y,test_size=.25)
model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=500)).fit(a,ya)
print(classification_report(yb,model.predict(b),digits=3))
```

## Laboratório aplicado revisado — Avaliação em conjunto de teste separado com baseline

Uma avaliação válida distingue treino, validação e teste e usa métricas pertinentes ao desbalanceamento, sem ajustar o modelo ao teste.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import balanced_accuracy_score, confusion_matrix, classification_report
X,y=make_classification(n_samples=240,n_features=8,n_informative=5,weights=[.75,.25],random_state=17)
Xt,Xe,yt,ye=train_test_split(X,y,test_size=.25,stratify=y,random_state=17)
base=DummyClassifier(strategy='most_frequent').fit(Xt,yt)
net=make_pipeline(StandardScaler(),MLPClassifier(hidden_layer_sizes=(12,),max_iter=400,random_state=17,early_stopping=True))
net.fit(Xt,yt)
for name,model in [('baseline',base),('MLP',net)]:
    pred=model.predict(Xe)
    print(name,'acurácia balanceada',round(balanced_accuracy_score(ye,pred),3),'matriz',confusion_matrix(ye,pred).tolist())
assert confusion_matrix(ye,net.predict(Xe)).shape==(2,2)
```

### Exportação de resultados (local)

Execute esta célula após o experimento acima. Os arquivos são salvos em `outputs/` e não entram automaticamente no Git.

```python
# Exportação — mesmas previsões e partição de teste do experimento.
from pathlib import Path
import pandas as pd
from sklearn.metrics import balanced_accuracy_score

out=Path('outputs'); out.mkdir(exist_ok=True)
linhas=[]
for nome,estimador in [('DummyClassifier',base),('MLPClassifier',net)]:
    pred=estimador.predict(Xe)
    linhas.append({'modelo':nome,'metrica':'balanced_accuracy_teste','valor':float(balanced_accuracy_score(ye,pred))})
caminho=out/'09_metricas_baselines.csv'
pd.DataFrame(linhas).to_csv(caminho,index=False)
print('Exportado:',caminho)
```
