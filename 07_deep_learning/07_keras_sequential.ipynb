Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
X,y=make_regression(n_samples=140,n_features=5,noise=10,random_state=17)
xt,xv,yt,yv=train_test_split(X,y,random_state=17)
base=Ridge().fit(xt,yt)
print('Baseline Ridge MAE:',round(mean_absolute_error(yv,base.predict(xv)),2))
```

## Laboratório aplicado revisado — MLP real com avaliação contra baseline em dados não vistos

A arquitetura Sequential aceita uma sequência linear de camadas, mas deve ser comparada com baseline e avaliada em dados não vistos.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error
X,y=make_regression(n_samples=220,n_features=5,noise=8,random_state=41)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=41)
base=DummyRegressor().fit(Xtr,ytr)
net=make_pipeline(StandardScaler(),MLPRegressor(
    hidden_layer_sizes=(32, 16),
    max_iter=1500,
    early_stopping=True,
    validation_fraction=0.15,
    n_iter_no_change=30,
    random_state=42
))
net.fit(Xtr,ytr)
mae_base=mean_absolute_error(yte,base.predict(Xte));mae_net=mean_absolute_error(yte,net.predict(Xte))
print('MAE baseline:',round(mae_base,2),'MAE MLP:',round(mae_net,2))
assert len(net.predict(Xte))==len(yte)
```
