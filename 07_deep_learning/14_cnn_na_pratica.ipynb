## Laboratório aplicado revisado — Classificador de imagens sintéticas com padrão aprendível

Uma CNN só demonstra aprendizado quando rótulos correspondem a padrões e o desempenho é medido fora do treino.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
rng=np.random.default_rng(14); n=260;X=np.zeros((n,12,12),dtype=np.float32);y=np.repeat([0,1],n//2)
for i in range(n):
    if y[i]==0:X[i,:,4:6]=1.0
    else:X[i,4:6,:]=1.0
    X[i]+=rng.normal(0,.22,size=(12,12))
idx=rng.permutation(n);X,y=X[idx],y[idx]
Xt,Xe,yt,ye=train_test_split(X.reshape(n,-1),y,test_size=.25,stratify=y,random_state=14)
m=LogisticRegression(max_iter=300).fit(Xt,yt)
a=accuracy_score(ye,m.predict(Xe))
print('Baseline linear para listras horizontais/verticais:',round(a,3));assert a>.8
```

### Interpretação, limites e verificação

Este experimento testa padrões com baseline linear. CNN verdadeira deve usar amostras de treino e teste separadas e métricas reais, não ruído com rótulos arbitrários.

**Checklist:** conferir entradas e saídas, sementes, inexistência de uso dos dados de teste durante ajuste e distinção entre uma demonstração sintética e desempenho em aplicações reais.

## Laboratório neural opcional — CNN treinada em dados com padrão

**Não executado nesta auditoria.** Quando TensorFlow estiver instalado, o código abaixo aprende listras horizontais ou verticais com `Conv2D`, em treino/teste separados. Sem TensorFlow, registra a dependência pendente.

```python
import importlib.util
if importlib.util.find_spec('tensorflow') is None:
    print('PENDENTE: instalar TensorFlow em ambiente compatível para o laboratório CNN.')
else:
    import numpy as np, tensorflow as tf
    from sklearn.model_selection import train_test_split
    tf.keras.utils.set_random_seed(14)
    rng=np.random.default_rng(14);X=np.zeros((160,12,12,1),dtype='float32');y=np.repeat([0,1],80)
    for i in range(len(X)):
        if y[i]==0:X[i,:,4:6,0]=1
        else:X[i,4:6,:,0]=1
        X[i]+=rng.normal(0,.2,size=X[i].shape)
    idx=rng.permutation(len(X));X,y=X[idx],y[idx]
    Xtr,Xte,ytr,yte=train_test_split(X,y,stratify=y,test_size=.25,random_state=14)
    model=tf.keras.Sequential([tf.keras.Input((12,12,1)),tf.keras.layers.Conv2D(6,3,activation='relu'),tf.keras.layers.GlobalAveragePooling2D(),tf.keras.layers.Dense(1,activation='sigmoid')])
    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
    history=model.fit(Xtr,ytr,epochs=12,batch_size=16,validation_split=.2,verbose=0)
    print('Acurácia teste CNN:',model.evaluate(Xte,yte,verbose=0)[1])
```
