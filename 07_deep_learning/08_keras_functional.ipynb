## Laboratório aplicado revisado — Dois caminhos de entrada e fusão de representações

Modelos Functional possuem grafo acíclico explícito e permitem entradas paralelas. A fusão NumPy é apenas explicação das dimensões.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
rng=np.random.default_rng(8)
a=rng.normal(size=(16,3));b=rng.normal(size=(16,2))
Wa=rng.normal(size=(3,4));Wb=rng.normal(size=(2,4))
ha=np.maximum(a@Wa,0);hb=np.maximum(b@Wb,0)
concat=np.concatenate([ha,hb],axis=1)
assert concat.shape==(16,8)
print('Entrada A:',a.shape,'Entrada B:',b.shape,'Fusão:',concat.shape)
print('Uma API Functional conecta duas entradas e une os tensores em um grafo.')
```
