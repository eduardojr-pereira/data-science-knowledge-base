## 2. Matemática essencial

**Expressão esquemática:** `out=⌊(n+2p−d(k−1)−1)/s⌋+1`

A expressão acima resume a relação central. Para a derivação, identifique dimensões, hipóteses e operações; não interprete notação compacta como uma implementação completa.

## 3. Experimento obrigatório — NumPy e/ou scikit-learn

Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
import numpy as np
im=np.arange(25).reshape(5,5)
k=np.array([[1.,0.],[-1.,0.]])
out=np.array([[np.sum(im[i:i+2,j:j+2]*k) for j in range(4)] for i in range(4)])
print('Imagem:',im.shape,'kernel:',k.shape,'saída valid:',out.shape)
assert out.shape==(4,4)
```

## Laboratório aplicado revisado — Convolução válida, padding e saída dimensional

Correlação cruzada, stride, padding e dilatação determinam o mapa de características; checar shapes e convenções.

**Procedimento:** execute a próxima célula com o núcleo declarado em `requirements.txt`, confira os `assert` e interprete os resultados. O experimento é independente dos exemplos anteriores.

```python
import numpy as np
image=np.arange(1,26,dtype=float).reshape(5,5)
filter=np.array([[1,0,-1],[1,0,-1],[1,0,-1]],dtype=float)
def corr_valid(X,K):
    h,w=K.shape
    return np.array([[np.sum(X[i:i+h,j:j+w]*K) for j in range(X.shape[1]-w+1)] for i in range(X.shape[0]-h+1)])
valid=corr_valid(image,filter)
padded=corr_valid(np.pad(image,1,mode='constant'),filter)
assert valid.shape==(3,3) and padded.shape==(5,5)
print('valid',valid.shape,'same via padding',padded.shape,'filtro vertical',valid[0].tolist())
```

### Interpretação, limites e verificação

Bibliotecas de CNN implementam correlação cruzada na prática; não confundir com convolução matemática que inverte o kernel.
