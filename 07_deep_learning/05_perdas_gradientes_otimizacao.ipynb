## 2. Matemática essencial

**Expressão esquemática:** `MSE=(1/n)Σ(ŷ−y)²`

A expressão acima resume a relação central. Para a derivação, identifique dimensões, hipóteses e operações; não interprete notação compacta como uma implementação completa.

## 3. Experimento obrigatório — NumPy e/ou scikit-learn

Este experimento é propositalmente pequeno: **executa sem GPU, internet ou serviço externo**. Quando há treino de baseline, não interpretá-lo como evidência da qualidade de uma rede neural.

```python
import numpy as np
x=np.array([1.,2.,3.]);y=np.array([2.,4.,6.]);w=0.
def mse(t): return np.mean((t*x-y)**2)
for _ in range(120):
 grad=2*np.mean((w*x-y)*x)
 w-=0.05*grad
eps=1e-5
num=(mse(w+eps)-mse(w-eps))/(2*eps)
ana=2*np.mean((w*x-y)*x)
print('w otimizado',round(w,5),'perda',round(mse(w),8),'gradiente erro',abs(num-ana))
assert abs(w-2)<1e-3
```

### Laboratório — descida do gradiente, diferença finita e convergência

Usaremos regressão linear com uma variável para conectar função de perda, derivada e otimização. Verificamos o gradiente pela diferença finita, acompanhamos a evolução da perda e distinguimos a métrica de treinamento de uma avaliação em dados reservados.

```python
import numpy as np

rng = np.random.default_rng(7)
x = np.linspace(-2., 2., 100)
y = 1.5 + 2.2*x + rng.normal(scale=.15, size=x.size)
indices = rng.permutation(x.size)
tr, te = indices[:80], indices[80:]
Xtr = np.column_stack([np.ones(tr.size), x[tr]])
Xte = np.column_stack([np.ones(te.size), x[te]])
ytr, yte = y[tr], y[te]

def perda(theta, X, alvo):
    return float(np.mean((X @ theta - alvo)**2))

def gradiente(theta, X, alvo):
    return (2 / len(alvo)) * X.T @ (X @ theta - alvo)

theta = np.array([0.1, -0.2])
h = 1e-6
aprox = np.array([(perda(theta + np.eye(2)[j]*h, Xtr, ytr) -
                   perda(theta - np.eye(2)[j]*h, Xtr, ytr))/(2*h) for j in range(2)])
assert np.allclose(aprox, gradiente(theta, Xtr, ytr), atol=1e-5)

historico = [perda(theta, Xtr, ytr)]
for _ in range(600):
    theta -= .1 * gradiente(theta, Xtr, ytr)
    historico.append(perda(theta, Xtr, ytr))
assert historico[-1] < historico[0]
print('Parâmetros estimados (intercepto, inclinação):', theta.round(3))
print('MSE inicial treino:', round(historico[0], 4))
print('MSE final treino:', round(historico[-1], 4))
print('MSE conjunto reservado:', round(perda(theta, Xte, yte), 4))
```

#### Análise e exercícios

**Interpretação:** o erro de diferença finita depende de `h`, e a convergência depende da taxa de aprendizado e da escala dos dados. Como são dados sintéticos, os resultados não demonstram generalização para situações reais. **Atividade:** teste taxas 0,01 e 0,5; compare curvas da perda e descreva quando a atualização se torna instável.
