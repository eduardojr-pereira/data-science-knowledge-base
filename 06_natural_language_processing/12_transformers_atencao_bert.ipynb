## Fundamentos conceituais

Transformers substituem recorrência estrita por operações de atenção e projeções, permitindo combinar informações de múltiplas posições. Em self-attention, consultas (Q), chaves (K) e valores (V) são transformações aprendidas de representações de tokens. BERT é um modelo de codificação bidirecional pré-treinado; seu treinamento e fine-tuning são diferentes de executar uma demonstração numérica de atenção.

## Modelo matemático e interpretação

Attention(Q,K,V)=softmax(QKᵀ/√d_k)V. O fator √d_k controla a escala do produto escalar. Em cenários autoregressivos, uma máscara causal bloqueia posições futuras.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
import numpy as np
Q=np.array([[1.,0.],[0.,1.],[1.,1.]])
K=Q.copy();V=np.array([[1.,2.],[2.,0.],[0.,1.]])
logits=Q@K.T/np.sqrt(K.shape[1]);scores=np.exp(logits-logits.max(axis=-1,keepdims=True));pesos=scores/scores.sum(axis=-1,keepdims=True)
resultado=pesos@V
print('Pesos de atenção:\n',pesos.round(3))
print('Representações contextualizadas:\n',resultado.round(3))
assert np.allclose(pesos.sum(axis=1),1)
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
mascara=np.triu(np.ones_like(logits,dtype=bool),k=1)
logits_m=np.where(mascara,-1e9,logits)
p=np.exp(logits_m-logits_m.max(axis=1,keepdims=True));p=p/p.sum(axis=1,keepdims=True)
print('Atenção causal (superior direita bloqueada):\n',p.round(3))
```

### Laboratório adicional — Atenção scaled dot-product em NumPy

Experimento independente do estado das células anteriores. Use-o como verificação e explique seus limites; resultados em conjuntos sintéticos não são estimativas de desempenho real.

```python
import numpy as np
q=np.array([[1.,0.],[0.,1.]])
k=np.array([[1.,0.],[0.,1.]])
v=np.array([[2.,1.],[1.,3.]])
s=q@k.T/np.sqrt(q.shape[-1]); e=np.exp(s-s.max(axis=-1,keepdims=True)); pesos=e/e.sum(axis=-1,keepdims=True)
print('Pesos (linhas somam 1):',np.round(pesos,3))
print('Saída contextualizada:',np.round(pesos@v,3))
```
