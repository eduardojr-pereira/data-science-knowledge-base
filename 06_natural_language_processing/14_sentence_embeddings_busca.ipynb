## Fundamentos conceituais

Embeddings de frases condensam informação de uma sequência em um vetor para busca por similaridade. A recuperação semântica pode aproximar paráfrases, mas também produzir falsos positivos. Neste núcleo obrigatório, utilizamos projeção densa de TF-IDF apenas como demonstração geométrica; ela NÃO equivale a um encoder neural de sentenças. O laboratório opcional descreve modelos Sentence Transformers.

## Modelo matemático e interpretação

Uma consulta vetorial q e documentos dᵢ podem ser ranqueados por sim(q,dᵢ). O índice de busca deve preservar dimensão, normalização e modelo de embedding para todos os registros.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity
frases=['A biblioteca empresta livros','Um aluno estuda na universidade','Livros podem ser consultados na biblioteca','Alunos fazem cursos de programação']
consulta='buscar livros da biblioteca'
v=TfidfVectorizer();M=v.fit_transform(frases+[consulta]);svd=TruncatedSVD(n_components=2,random_state=7);E=svd.fit_transform(M)
scores=cosine_similarity(E[-1:],E[:-1]).ravel()
print('Ranking ilustrativo (LSA de TF-IDF):')
for i in scores.argsort()[::-1]:print(round(float(scores[i]),3),frases[i])
```
