## Fundamentos conceituais

Uma linguagem natural possui níveis de análise: morfológico (formas de palavras), sintático (relações estruturais), semântico (significados) e pragmático (efeitos de contexto). Um mesmo texto pode conter ambiguidades que um conjunto de regras fixas não resolve. PLN não equivale a apenas remover pontuação: o pré-processamento deve respeitar a tarefa e o significado relevante.

## Modelo matemático e interpretação

Um corpus é representado como D = {d₁, …, dₙ}. O pipeline de aprendizagem transforma cada documento dᵢ em características xᵢ e, quando há rótulo yᵢ, ajusta uma função f(xᵢ) para aproximar yᵢ. Aprendizagem generativa estima distribuições sobre sequências de tokens, não somente rótulos.
