## Fundamentos conceituais

Em classificação de textos, o vectorizer precisa ser ajustado apenas nos dados de treinamento. A combinação de TF-IDF com regressão logística ou Naive Bayes fornece baselines fortes para coleções pequenas. Uma análise de sentimentos depende de contexto, domínio e rótulos confiáveis; o modelo aprende padrões presentes no conjunto, não o sentimento verdadeiro de qualquer frase.

## Modelo matemático e interpretação

Uma função de decisão linear possui score f(x)=wᵀx+b; a regressão logística usa σ(f)=1/(1+e^{−f}). As métricas precision, recall e F1 complementam acurácia, principalmente com classes desbalanceadas.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import LeaveOneOut,cross_val_predict
from sklearn.metrics import confusion_matrix,classification_report
positivos=['atendimento ótimo e rápido','produto excelente e eficiente','suporte gentil e prestativo','experiência muito boa','serviço excelente','compra satisfatória']
negativos=['atendimento péssimo e demorado','produto ruim e quebrado','suporte grosseiro e lento','experiência muito ruim','serviço péssimo','compra insatisfatória']
X=positivos+negativos;y=[1]*len(positivos)+[0]*len(negativos)
pipe=Pipeline([('tfidf',TfidfVectorizer(ngram_range=(1,2))),('modelo',LogisticRegression(max_iter=1000))])
pred=cross_val_predict(pipe,X,y,cv=LeaveOneOut())
print('Matriz de confusão:\n',confusion_matrix(y,pred))
print(classification_report(y,pred,zero_division=0))
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
pipe.fit(X,y)
for frase in ['atendimento excelente','serviço ruim']:
    print(frase,'->',int(pipe.predict([frase])[0]))
```

### Laboratório adicional — Partição estratificada de texto antes da vetorização

Experimento independente do estado das células anteriores. Use-o como verificação e explique seus limites; resultados em conjuntos sintéticos não são estimativas de desempenho real.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
X=['bom ótimo', 'excelente bom','ruim péssimo','horrível ruim','gostei muito','não gostei','ótimo resultado','resultado ruim']
y=[1,1,0,0,1,0,1,0]
Xt,Xv,yt,yv=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
m=make_pipeline(TfidfVectorizer(), LogisticRegression(max_iter=200)).fit(Xt,yt)
print('Acurácia ilustrativa (amostra pequena):',m.score(Xv,yv))
```
