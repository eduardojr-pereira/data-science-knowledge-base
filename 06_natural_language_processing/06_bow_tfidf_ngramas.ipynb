## Fundamentos conceituais

Bag-of-Words ignora a ordem global e usa frequências de termos; n-gramas recuperam dependências locais ao custo de maior dimensionalidade. TF-IDF reduz o peso de termos presentes em muitos documentos, mas isso não implica compreensão semântica. A frequência depende da preparação e do vocabulário fixado no treino.

## Modelo matemático e interpretação

Uma forma comum de IDF suavizado é idf(t)=log((1+n)/(1+df(t)))+1. O peso TF-IDF é a frequência do termo multiplicada pela importância inversa do documento, com variantes de normalização.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
corpus=['dados e ciência','ciência de dados','linguagem natural com dados']
bow=CountVectorizer(); X=bow.fit_transform(corpus)
print('Vocabulário:',bow.get_feature_names_out().tolist())
print('Contagem:',X.toarray())
tfidf=TfidfVectorizer();T=tfidf.fit_transform(corpus)
print('TF-IDF arredondado:',T.toarray().round(2))
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
bigramas=CountVectorizer(ngram_range=(1,2)).fit(corpus)
print('Exemplos de bigramas:',[t for t in bigramas.get_feature_names_out() if ' ' in t][:8])
```
