## Fundamentos conceituais

Retrieval-Augmented Generation (RAG) combina recuperação de documentos com geração condicionada a evidências. Um sistema robusto deve registrar fontes, identificar quando faltam evidências e não interpretar todo texto recuperado como instrução confiável. A demonstração obrigatória é somente recuperação com resposta baseada em trechos literais — sem LLM externo.

## Modelo matemático e interpretação

O recuperador ranqueia trechos cᵢ com score sim(q,cᵢ); o gerador condicionado estima P(resposta | pergunta, trechos selecionados). A qualidade do sistema depende das duas etapas, que devem ser avaliadas separadamente.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
base=[('doc01','A biblioteca funciona de segunda a sexta-feira.'),('doc02','O laboratório de informática oferece computadores para estudo.'),('doc03','O empréstimo de livros exige cadastro válido.')]
pergunta='Quando a biblioteca funciona?'
v=TfidfVectorizer();X=v.fit_transform([t for _,t in base]);q=v.transform([pergunta]);scores=cosine_similarity(q,X).ravel()
pos=int(scores.argmax())
print('Pergunta:',pergunta)
print('Trecho recuperado:',base[pos], '| score=',round(float(scores[pos]),3))
print('Resposta extrativa:',base[pos][1])
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
limiar=0.15
pergunta_nova='Qual é a previsão do tempo amanhã?'
q=v.transform([pergunta_nova]);scores=cosine_similarity(q,X).ravel()
print('Sem evidência suficiente' if float(scores.max())<limiar else 'Verificar manualmente a relevância')
```
