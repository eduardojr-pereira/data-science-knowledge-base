## Fundamentos conceituais

Sumarização extrativa escolhe frases existentes no documento, em contraste com uma sumarização abstrativa que produz frases novas. Métodos de frequência priorizam sentenças com termos salientes. A heurística clássica de Luhn usa proximidade de palavras significativas; neste capítulo implementamos uma aproximação pedagógica explícita, não uma reprodução literal de todas as variantes históricas.

## Modelo matemático e interpretação

Um score simples é S(s)=Σ_{t em s} f(t) / |s|. Para uma janela de Luhn, uma pontuação típica é (nº de termos significativos)² / (extensão da janela), que favorece concentração.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
import re
from collections import Counter
texto=('A biblioteca abriu novas salas de estudo. '
       'Estudantes consultaram livros e artigos científicos. '
       'As novas salas receberam computadores. '
       'A universidade planeja ampliar os horários de acesso.')
frases=[s.strip() for s in re.split(r'(?<=[.!?])\s+',texto) if s.strip()]
tok=lambda s:re.findall(r'\b\w+\b',s.lower())
stop={'a','as','de','e','os','do','da','o','em'}
freq=Counter(t for s in frases for t in tok(s) if t not in stop)
scores=[sum(freq[t] for t in tok(s) if t not in stop)/max(len(tok(s)),1) for s in frases]
indices=sorted(sorted(range(len(frases)),key=lambda i:scores[i],reverse=True)[:2])
print('Resumo:\n'+' '.join(frases[i] for i in indices))
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
def score_luhn_aproximado(frase, relevantes):
    termos=tok(frase)
    pos=[i for i,t in enumerate(termos) if t in relevantes]
    return len(pos)**2/(pos[-1]-pos[0]+1) if pos else 0
relevantes=set(t for t,n in freq.most_common(6))
print('Scores Luhn aproximados:',[round(score_luhn_aproximado(s,relevantes),2) for s in frases])
```
