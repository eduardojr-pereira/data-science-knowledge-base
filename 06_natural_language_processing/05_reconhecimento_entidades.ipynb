## Fundamentos conceituais

Reconhecimento de entidades nomeadas (NER/REN) tenta localizar menções como pessoas, organizações e localidades e atribuir um tipo. A simples presença de letra maiúscula não basta: início de frase e nomes próprios coexistem. Em extração de informação, erros de fronteira podem ser tão graves quanto erros de classificação.

## Modelo matemático e interpretação

Precisão = TP/(TP+FP) e recall = TP/(TP+FN) para uma política definida de comparação das entidades previstas e anotadas. Em correspondência exata, início, fim e tipo devem concordar.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
import re
texto='Ana visitou a Universidade Federal em Belo Horizonte.'
padroes={'Pessoa':r'\bAna\b','Organização':r'Universidade Federal','Local':r'Belo Horizonte'}
extracoes=[(tipo,m.group(),m.span()) for tipo,p in padroes.items() for m in re.finditer(p,texto)]
print(extracoes)
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
ouro={('Ana','Pessoa'),('Universidade Federal','Organização'),('Belo Horizonte','Local')}
previstas={('Ana','Pessoa'),('Belo Horizonte','Local'),('Federal','Organização')}
tp=len(ouro & previstas);precision=tp/len(previstas);recall=tp/len(ouro)
print({'precision':round(precision,3),'recall':round(recall,3)})
```
