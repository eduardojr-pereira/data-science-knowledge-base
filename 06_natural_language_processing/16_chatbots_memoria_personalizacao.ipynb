## Fundamentos conceituais

Sistemas conversacionais precisam separar estado da tarefa, histórico recente, preferências explícitas e memória duradoura. Uma mudança de estilo (“seja objetivo”) não deve modificar a evidência nem as regras de segurança. O gerenciamento de diálogo inclui acompanhamento do objetivo e estratégias para solicitar esclarecimentos. Memória persistente exige consentimento, minimização, retenção e possibilidade de correção e exclusão.

## Modelo matemático e interpretação

Um estado conversacional s_t pode ser atualizado com s_{t+1}=F(s_t, mensagem_t). A política de resposta π(resposta | s_t, contexto) considera preferências, mas a veracidade e as restrições não devem depender do grau de informalidade.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from dataclasses import dataclass,field
@dataclass
class EstadoDialogo:
    objetivo:str='explicar conceito'
    tom:str='neutro'
    mensagens:list=field(default_factory=list)
def explicar_knn(estado):
    base='KNN classifica usando os exemplos mais próximos no espaço de características.'
    if estado.tom=='informal':return 'Bora entender: '+base
    if estado.tom=='resumido':return 'KNN usa vizinhos próximos para classificar.'
    return base
for tom in ['neutro','informal','resumido']:
    e=EstadoDialogo(tom=tom)
    print(tom,'->',explicar_knn(e))
```

```python
historico=[]
def registrar_turno(papel,texto,limite=4):
    historico.append({'papel':papel,'texto':texto})
    del historico[:-limite]
registrar_turno('usuario','Explique KNN');registrar_turno('assistente','KNN usa distâncias.')
registrar_turno('usuario','Mais exemplos, por favor.')
print('Histórico recente (somente memória em sessão):',historico)
```
