## Fundamentos conceituais

Stemming reduz formas lexicais usando regras e pode gerar radicais que não são palavras. Lematização depende de conhecimento morfológico para obter formas canônicas. Etiquetagem morfossintática (POS tagging) identifica classes como substantivo, verbo e adjetivo em função do contexto. O erro comum é tratar stemming como uma forma imprecisa de lematização: são objetivos diferentes.

## Modelo matemático e interpretação

A etiqueta morfossintática é prevista condicionalmente ao contexto: classeᵢ = argmax_c P(c | tokens e contexto). Em modelos estatísticos, esse problema é uma classificação sequencial.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from nltk.stem.snowball import SnowballStemmer
stemmer=SnowballStemmer('portuguese')
palavras=['estudando','estudantes','estudos','pesquisadores','pesquisando']
for p in palavras: print(f'{p:16s} -> {stemmer.stem(p)}')
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.
