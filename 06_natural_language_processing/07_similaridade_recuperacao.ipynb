## Fundamentos conceituais

A recuperação clássica representa consulta e documentos no mesmo espaço de termos. A similaridade por cosseno compara orientação, não o comprimento, e funciona particularmente bem com representações normalizadas e esparsas. BM25 é outra abordagem lexical relevante, mas sua implementação detalhada não é necessária para aprender TF-IDF.

## Modelo matemático e interpretação

sim_cosseno(a,b)=(a·b)/(||a|| ||b||), definida para vetores não nulos. Uma representação TF-IDF normalizada utiliza produto escalar como similaridade de cosseno.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
corpus=['A biblioteca tem livros de dados','O banco de dados está atualizado','A universidade ensina programação em Python']
consulta='livros e biblioteca'
v=TfidfVectorizer();D=v.fit_transform(corpus);q=v.transform([consulta]);scores=cosine_similarity(q,D).ravel()
for i in scores.argsort()[::-1]: print(f'{scores[i]:.3f} | {corpus[i]}')
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
from numpy.linalg import norm
import numpy as np
a=np.array([1.,1.,0.]);b=np.array([2.,2.,0.]);c=np.array([0.,0.,1.])
print('Similaridade a-b:',np.dot(a,b)/(norm(a)*norm(b)))
print('Similaridade a-c:',np.dot(a,c)/(norm(a)*norm(c)))
```

### Laboratório adicional — Recuperação lexical com escores observáveis

Experimento independente do estado das células anteriores. Use-o como verificação e explique seus limites; resultados em conjuntos sintéticos não são estimativas de desempenho real.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
textos=['custo do frete rodoviário','prazo de entrega rápida','frete e entrega de mercadoria']
v=TfidfVectorizer()
mat=v.fit_transform(textos)
query=v.transform(['frete de mercadoria'])
print(sorted(enumerate(cosine_similarity(query,mat).ravel().round(3)),key=lambda x:-x[1]))
```
