## Fundamentos conceituais

Embeddings distribucionais representam palavras como vetores densos aprendidos a partir de contextos. Word2Vec aprende relações usando janelas de contexto; não equivale a um dicionário explícito de sinônimos. Embeddings estáticos atribuem normalmente um vetor por token, enquanto modelos contextuais podem representar a mesma palavra diferentemente conforme a frase.

## Modelo matemático e interpretação

Na abordagem Skip-gram, o objetivo é maximizar log P(contexto | palavra central). Uma aproximação por matriz termo-contexto e redução dimensional ilustra a hipótese distribucional, sem reproduzir o treinamento neural do Word2Vec.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import TruncatedSVD
frases=['gato dorme na casa','cachorro dorme na casa','gato brinca no quintal','cachorro brinca no quintal','aluno estuda ciência','professor ensina ciência']
V=CountVectorizer();X=V.fit_transform(frases)
Z=TruncatedSVD(n_components=2,random_state=42).fit_transform(X)
print('Representação densa ilustrativa por documento (SVD, não Word2Vec):')
for frase,vetor in zip(frases,Z):print(frase,vetor.round(3))
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
import numpy as np
print('Dimensionalidades: contagem esparsa =',X.shape[1],'; representação densa =',Z.shape[1])
assert np.isfinite(Z).all()
```
