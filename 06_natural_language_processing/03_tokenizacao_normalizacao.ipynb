## Fundamentos conceituais

Tokens não são necessariamente palavras. Uma tokenização subword pode decompor uma palavra em partes, enquanto tokenizadores de palavras tentam respeitar pontuação e contrações. Stopwords são palavras frequentes em um corpus; retirar automaticamente “não” pode inverter a polaridade de sentimentos. Uma regra de normalização precisa ser avaliada junto à tarefa final.

## Modelo matemático e interpretação

A frequência relativa de um token w em um documento d é f(w,d)/|d|, com |d| representando o número de tokens. Essa estatística é sensível ao tokenizador e às regras de filtragem.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
import re
frase='Eu NÃO gostei do atendimento, mas gostei do produto!'
tokens=re.findall(r'\b\w+\b',frase.lower(),flags=re.UNICODE)
stopwords={'eu','do','mas'}
print('Tokens:',tokens)
print('Após filtragem controlada:',[t for t in tokens if t not in stopwords])
assert 'não' in tokens
```

**Interpretação guiada:** verifique se o resultado corresponde à fórmula e à hipótese discutidas. Anote o que muda com a entrada e quais conclusões **não** podem ser generalizadas a partir deste exemplo.

```python
from collections import Counter
texto='Gosto de ciência de dados e gosto de aprender PLN.'
tokens=re.findall(r'\b\w+\b',texto.casefold())
print(Counter(tokens).most_common(5))
```
