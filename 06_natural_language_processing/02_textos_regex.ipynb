## Fundamentos conceituais

Expressões regulares reconhecem padrões de caracteres, mas não interpretam o significado de um documento. Para datas, números, emails e pontuação, defina explicitamente quais elementos devem permanecer. A normalização deve ter objetivos mensuráveis. Minúsculas podem ajudar em certas tarefas, mas podem prejudicar reconhecimento de entidades.

## Modelo matemático e interpretação

Uma transformação de texto T: Σ* → Σ* transforma cadeias de símbolos; normalmente T não é inversível. Isso significa perda de informação, que deve ser justificada. Em classificação, a transformação aplicada em treino e inferência deve ser idêntica.

## Implementação reproduzível

O código abaixo usa textos sintéticos ou bibliotecas locais. **Não executa downloads nem acessa APIs externas.** Exemplos opcionais só rodam quando a dependência já está disponível; a ausência deles não interrompe o notebook.

```python
import re
frase = '  Olá,   Brasil! Contato: exemplo@dominio.org; valor R$ 1.250,00.  '
frase_limpa = re.sub(r'\s+', ' ', frase).strip()
contatos = re.findall(r'[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}', frase_limpa)
print(frase_limpa, '\nEmails:', contatos)
```

### Laboratório adicional — Regex: diferença entre encontrar e validar

Experimento independente do estado das células anteriores. Use-o como verificação e explique seus limites; resultados em conjuntos sintéticos não são estimativas de desempenho real.

```python
import re
amostras = ['cliente@exemplo.com', 'cliente@@exemplo.com', ' 123 ', 'texto 123 outro']
for s in amostras:
    print(repr(s), 'contém número=',bool(re.search(r'\d+',s)),
          'é somente número=',bool(re.fullmatch(r'\s*\d+\s*',s)))
```
