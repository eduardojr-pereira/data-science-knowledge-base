## Fundamentos conceituais

Um pixel não é um objeto: imagens são medições discretas da luz. O problema visual depende de aquisição, domínio, rótulos, resolução, iluminação e custo de erro. Um pipeline inclui aquisição → pré-processamento → características → inferência → avaliação. Distinga classificação de imagem inteira, detecção de objetos com caixas, segmentação por pixel e reconhecimento de indivíduos.

### Conceito matemático

$$I\in\mathbb{R}^{H\times W\times C}$$

A fórmula expressa o mecanismo principal do capítulo. Identifique unidades, dimensões e hipóteses antes de aplicar uma implementação.

## Experimento 1 — construção e medidas

Execute em kernel limpo. Exemplos usam datasets de bibliotecas instaláveis e não requerem download de arquivos externos.

```python
from sklearn.datasets import load_digits
import numpy as np
D=load_digits()
I=D.images.astype('float32')/16
print('n imagens:',len(I),'shape de cada:',I.shape[1:],'classes:',len(np.unique(D.target)))
assert I.min()>=0 and I.max()<=1
```

## Experimento 2 — visualização, comparação ou validação

Observe não apenas o valor obtido, mas o que o experimento **realmente demonstra**.

```python
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,5,figsize=(10,2))
for ax,i in zip(axes,range(5)):
 ax.imshow(I[i],cmap='gray',vmin=0,vmax=1);ax.set_title(f'Dígito {D.target[i]}');ax.axis('off')
plt.tight_layout();plt.show()
```
