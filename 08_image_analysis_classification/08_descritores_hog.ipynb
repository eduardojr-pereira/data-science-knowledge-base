## Fundamentos conceituais

HOG representa distribuições locais de orientações de gradientes. Divide a imagem em células, agrega orientações em histogramas e normaliza blocos para reduzir sensibilidade ao contraste. Funciona bem em certas condições estruturadas, mas é limitado por pose, escala e oclusões. HOG é uma característica, não um detector isoladamente.

### Conceito matemático

$$\theta=\operatorname{atan2}(G_y,G_x)$$

A fórmula expressa o mecanismo principal do capítulo. Identifique unidades, dimensões e hipóteses antes de aplicar uma implementação.

## Experimento 1 — construção e medidas

Execute em kernel limpo. Exemplos usam datasets de bibliotecas instaláveis e não requerem download de arquivos externos.

```python
from sklearn.datasets import load_digits
from skimage.feature import hog
import numpy as np
D=load_digits();feats=[]
for im in D.images:
 feats.append(hog(im,pixels_per_cell=(4,4),cells_per_block=(1,1),orientations=9))
X=np.asarray(feats)
print('HOG shape:',X.shape,'valores finitos:',np.isfinite(X).all())
```

## Experimento 2 — visualização, comparação ou validação

Observe não apenas o valor obtido, mas o que o experimento **realmente demonstra**.

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score
xtr,xte,ytr,yte=train_test_split(X,D.target,test_size=.25,random_state=42,stratify=D.target)
clf=LinearSVC(dual=False,max_iter=8000).fit(xtr,ytr)
print('HOG + LinearSVC (digits):',round(accuracy_score(yte,clf.predict(xte)),3))
```

## Interpretação técnica

O experimento mostra classificação de dígitos com HOG; não demonstra detecção de objetos em cenas.
