## CAP sob partição de rede

Em presença de uma **partição de rede**, a aplicação enfrenta trade-off entre disponibilidade das requisições e consistência forte (linearizabilidade). Não é uma escolha universal de quaisquer duas letras em operação normal.

```python
print('Pergunta de projeto: durante uma partição, podemos rejeitar operações para preservar consistência?')
```

## Consistência eventual e replicação

Replicar dados não garante resposta imediatamente atualizada em todos os nós. Consistência de transação ACID difere da propriedade de consistência do CAP.

```python
cenarios=pd.DataFrame({'cenario':['Saldo financeiro','Curtidas em rede social','Log de sensores'],'preocupacao':['invariantes fortes','atualização possivelmente assíncrona','throughput de escrita']});display(cenarios)
```

## Exercício resolvido — decisão arquitetural

Para um sistema de pontos acumulados, tolerar temporariamente uma leitura desatualizada pode ser aceitável; para impedir gasto duplo, exige-se um desenho com invariantes fortes. Justifique por operação, não apenas por rótulo do banco.
