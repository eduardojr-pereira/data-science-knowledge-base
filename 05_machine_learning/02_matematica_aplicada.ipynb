## 1. Fundamentos, objetivos e contexto

**Objetivos.** Usar vetores, matrizes, normas, distâncias, projeções, gradientes e regularização em problemas concretos.

Retomada de NumPy (Disciplina 01) e probabilidade (Disciplina 03), sem refazer seus capítulos. Produto escalar mede alinhamento; norma L2 mede comprimento; distância euclidiana é sensível à escala. Gradientes indicam direção de aumento local de uma função diferenciável; gradiente descendente atualiza parâmetros na direção oposta. Autovetores/autovalores serão retomados em PCA.


## 2. Fundamentos matemáticos

$\lVert x\rVert_2=\sqrt{\sum_j x_j^2}$, $d_2(x,z)=\lVert x-z\rVert_2$ e $x^\top z=\sum_jx_jz_j$.

Para $J(w)=\frac1n\sum_i (x_i^\top w-y_i)^2$, $\nabla J(w)=\frac2n X^\top(Xw-y)$. Com Ridge: $J_\lambda(w)=J(w)+\lambda\lVert w\rVert_2^2$ (intercepto não penalizado na prática).

## 3. Experimento guiado: preparando a pergunta

**Antes do código**, identifique: (a) quais dados são de treino, (b) qual informação poderá entrar no modelo, (c) que métrica ou propriedade será observada, e (d) que conclusão não é permitida pelos resultados. Estes exemplos são didáticos, com semente fixa onde aplicável.


```python
import numpy as np
x=np.array([2.,1.]); z=np.array([4.,5.]);print('Produto escalar:',x@z,'normas:',np.linalg.norm(x),np.linalg.norm(z))
print('Distância:',np.linalg.norm(x-z))
A=np.array([[2.,0.],[0.,1.]])
values,vectors=np.linalg.eigh(A)
print('Autovalores:',values,'autovetores (colunas):\n',vectors)
```

## 4. Segundo experimento ou extensão controlada

Mude apenas uma condição principal (algoritmo, parâmetro ou tipo de dado), quando possível, e registre o efeito. Não ajuste decisões com base no teste final.


```python
import numpy as np
rng=np.random.default_rng(42)
X=np.c_[np.ones(100),rng.normal(size=100)];y=1.5+2.3*X[:,1]+rng.normal(scale=.25,size=100)
w=np.zeros(2);learning_rate=.15
loss=[]
for step in range(200):
    residual=X@w-y
    loss.append(np.mean(residual**2))
    grad=2*X.T@residual/len(y)
    w-=learning_rate*grad
print('Parâmetros aprendidos:',w.round(3),'MSE inicial/final:',round(loss[0],3),round(loss[-1],3))
```

## 5. Exercício resolvido

**Enunciado:** reforce o conceito central com um exemplo menor, cujo resultado pode ser inspecionado diretamente.


```python
# Exercício resolvido: comparar distâncias antes e depois de escalonar
from sklearn.preprocessing import StandardScaler
X=np.array([[20.,2000.],[22.,2500.],[25.,3000.]])
print('Distância bruta:',round(np.linalg.norm(X[0]-X[1]),2))
Xs=StandardScaler().fit_transform(X) # Apenas ilustração matemática, NÃO avaliação preditiva
print('Distância padronizada:',round(np.linalg.norm(Xs[0]-Xs[1]),2))
```

## 6. Interpretação, limitações e erros comuns

O gradiente descendente demonstrado é uma implementação pedagógica de mínimos quadrados, não um otimizador universal. Sem normalização, um atributo de escala enorme pode dominar uma distância. Em modelos preditivos, o escalonador deve ser ajustado APENAS no treino.
