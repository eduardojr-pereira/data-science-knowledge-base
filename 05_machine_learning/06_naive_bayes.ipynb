## 1. Fundamentos, objetivos e contexto

**Objetivos.** Transformar o Teorema de Bayes (Disciplina 03 — NB09) em um classificador, compreender independência condicional e diferenças entre GaussianNB, MultinomialNB e BernoulliNB.

Naive Bayes assume independência **condicional à classe**; isso é uma simplificação geralmente não verdadeira, mas pode resultar em boa classificação. `GaussianNB` modela atributos contínuos por gaussianas; `MultinomialNB` é comum para contagens; `BernoulliNB`, para atributos binários. A qualidade da decisão e a calibração probabilística são questões distintas.


## 2. Fundamentos matemáticos

$P(y\mid x_1,\ldots,x_p)\propto P(y)\prod_{j=1}^pP(x_j\mid y)$. O classificador escolhe a classe com maior produto (na implementação, somam-se log-probabilidades por estabilidade numérica). Em `GaussianNB`, cada $P(x_j\mid y)$ segue uma densidade normal estimada por classe.

**Conexão com disciplinas anteriores:** revise os conceitos matemáticos e estatísticos correspondentes. Não repetimos integralmente os capítulos de Python, visualização e inferência estatística.


## 3. Experimento guiado: preparando a pergunta

**Antes do código**, identifique: (a) quais dados são de treino, (b) qual informação poderá entrar no modelo, (c) que métrica ou propriedade será observada, e (d) que conclusão não é permitida pelos resultados. Estes exemplos são didáticos, com semente fixa onde aplicável.


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report,confusion_matrix
X,y=load_iris(return_X_y=True)
a,b,c,d=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
clf=GaussianNB().fit(a,c)
print('Prioris aprendidas:',clf.class_prior_.round(3))
print('Matriz de confusão:\n',confusion_matrix(d,clf.predict(b)))
print(classification_report(d,clf.predict(b),zero_division=0))
```

## 4. Segundo experimento ou extensão controlada

Mude apenas uma condição principal (algoritmo, parâmetro ou tipo de dado), quando possível, e registre o efeito. Não ajuste decisões com base no teste final.


```python
import numpy as np
from sklearn.naive_bayes import MultinomialNB,BernoulliNB
C=np.array([[2,0],[1,0],[0,2],[0,3],[2,1],[0,1]])
y2=np.array([0,0,1,1,0,1])
mult=MultinomialNB(alpha=1).fit(C,y2)
bern=BernoulliNB(alpha=1).fit((C>0).astype(int),y2)
print('Multinomial:',mult.predict([[1,0]]),'Bernoulli:',bern.predict([[1,0]]))
```
