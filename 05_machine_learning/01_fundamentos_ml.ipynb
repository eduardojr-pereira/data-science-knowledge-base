## 1. Fundamentos, objetivos e contexto

**Objetivos.** Distinguir tarefas de classificação, regressão e agrupamento; reconhecer aprendizado supervisionado/não supervisionado, *batch*/*online*, baseado em instâncias/modelos; formular objetivo e baseline.

**Panorama:** um modelo aprende parâmetros ou estrutura a partir de exemplos; o objetivo preditivo é generalizar para observações não vistas. Não confundir ajuste no treino com qualidade fora da amostra. Métodos baseados em instâncias (ex.: KNN) usam exemplos para decidir; baseados em modelos (ex.: regressão logística) aprendem parâmetros. Em *batch*, reentrena-se em lotes; em *online*, atualiza-se incrementalmente quando o estimador dá suporte a isso.

**Outros paradigmas:** regras de associação procuram coocorrências, não se confundem com classificação. Aprendizado por reforço é apenas contextualizado: não faz parte do escopo prático desta disciplina.


## 2. Fundamentos matemáticos

**Notação:** dados $X\in\mathbb{R}^{n\times p}$, rótulos $y\in\mathcal{Y}^n$ e preditor $\hat y=f_\theta(X)$. O risco de generalização é $R(f)=\mathbb{E}_{(X,Y)\sim P}[L(Y,f(X))]$, desconhecido; a avaliação em dados separados aproxima esse risco sob hipóteses sobre a amostragem.

**Conexão com disciplinas anteriores:** revise os conceitos matemáticos e estatísticos correspondentes. Não repetimos integralmente os capítulos de Python, visualização e inferência estatística.


## 3. Experimento guiado: preparando a pergunta

**Antes do código**, identifique: (a) quais dados são de treino, (b) qual informação poderá entrar no modelo, (c) que métrica ou propriedade será observada, e (d) que conclusão não é permitida pelos resultados. Estes exemplos são didáticos, com semente fixa onde aplicável.


```python
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier, DummyRegressor
from sklearn.metrics import accuracy_score, mean_absolute_error
Xc, yc = load_iris(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(Xc, yc, test_size=.25, random_state=42, stratify=yc)
clf_base = DummyClassifier(strategy='most_frequent').fit(Xtr,ytr)
print('Baseline classificação (accuracy):',round(accuracy_score(yte,clf_base.predict(Xte)),3))
Xr,yr = load_diabetes(return_X_y=True)
Xtr_r,Xte_r,ytr_r,yte_r=train_test_split(Xr,yr,test_size=.25,random_state=42)
reg_base=DummyRegressor(strategy='mean').fit(Xtr_r,ytr_r)
print('Baseline regressão (MAE):',round(mean_absolute_error(yte_r,reg_base.predict(Xte_r)),2))
```

### Laboratório adicional — Baseline antes do algoritmo

Experimento independente do estado das células anteriores. Use-o como verificação e explique seus limites; resultados em conjuntos sintéticos não são estimativas de desempenho real.


```python
import numpy as np
from sklearn.dummy import DummyClassifier
from sklearn.metrics import balanced_accuracy_score
X = np.arange(12).reshape(-1, 1)
y = np.array([0]*9 + [1]*3)
m = DummyClassifier(strategy="most_frequent").fit(X[:8], y[:8])
print("Distribuição observada:", np.unique(y, return_counts=True))
print("Predições do baseline:", m.predict(X[8:]).tolist())
print("Acurácia balanceada:", round(balanced_accuracy_score(y[8:], m.predict(X[8:])), 3))
```
