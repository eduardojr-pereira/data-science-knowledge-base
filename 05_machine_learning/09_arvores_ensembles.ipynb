## 1. Fundamentos, objetivos e contexto

**Objetivos.** Explicar divisão recursiva, Gini/entropia, poda, bagging, Random Forest e boosting; comparar classificação e regressão por árvores.

Árvores podem capturar interações e não linearidades sem escalonamento obrigatório. Árvores profundas são instáveis; Random Forest combina árvores treinadas sobre amostras e subconjuntos aleatórios de atributos. Boosting combina estimadores sequenciais que corrigem erros do conjunto. Importâncias baseadas em impureza podem favorecer atributos com muitas possibilidades de corte.


## 2. Fundamentos matemáticos

Impureza de Gini $G=1-\sum_k p_k^2$ e entropia $H=-\sum_k p_k\log_2p_k$. Para regressão, um critério usual é a redução de erro quadrático. No bagging, $\hat f(x)=\frac1B\sum_{b=1}^B f_b(x)$ para regressão (ou votação para classificação).

**Conexão com disciplinas anteriores:** revise os conceitos matemáticos e estatísticos correspondentes. Não repetimos integralmente os capítulos de Python, visualização e inferência estatística.


## 3. Experimento guiado: preparando a pergunta

**Antes do código**, identifique: (a) quais dados são de treino, (b) qual informação poderá entrar no modelo, (c) que métrica ou propriedade será observada, e (d) que conclusão não é permitida pelos resultados. Estes exemplos são didáticos, com semente fixa onde aplicável.


```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,HistGradientBoostingClassifier
from sklearn.metrics import balanced_accuracy_score
X,y=load_wine(return_X_y=True);a,b,c,d=train_test_split(X,y,stratify=y,test_size=.3,random_state=42)
models={'árvore':DecisionTreeClassifier(max_depth=4,random_state=42),'forest':RandomForestClassifier(n_estimators=100,random_state=42),'boosting':HistGradientBoostingClassifier(max_iter=70,random_state=42)}
for name,m in models.items():
    m.fit(a,c);print(name,'balanced_accuracy',round(balanced_accuracy_score(d,m.predict(b)),3))
```

### Como interpretar

Os números são resultados específicos do experimento e da amostra, não garantias universais. Verifique o que mudou, qual método foi usado e qual é o cenário de aplicação. Compare com a teoria matemática e não confunda desempenho de treino, validação e teste.


## 4. Segundo experimento ou extensão controlada

Mude apenas uma condição principal (algoritmo, parâmetro ou tipo de dado), quando possível, e registre o efeito. Não ajuste decisões com base no teste final.


```python
from sklearn.datasets import load_diabetes
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
X,y=load_diabetes(return_X_y=True);a,b,c,d=train_test_split(X,y,test_size=.25,random_state=42)
for m in [DecisionTreeRegressor(max_depth=3,random_state=42),RandomForestRegressor(n_estimators=80,random_state=42),HistGradientBoostingRegressor(max_iter=60,random_state=42)]:
    m.fit(a,c);print(type(m).__name__,'MAE',round(mean_absolute_error(d,m.predict(b)),2))
```
