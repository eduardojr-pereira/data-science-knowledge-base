## 1. Fundamentos, objetivos e contexto

**Objetivos.** Comparar transformações polinomiais, SVR, árvores e Random Forest em regressão; interpretar MAE, MSE, RMSE, R² e resíduos.

Regressão polinomial é linear nos coeficientes após a transformação dos atributos. Transformação de grau alto favorece sobreajuste. SVR controla tolerância $\epsilon$ e regularização $C$. Para séries temporais, não misturar observações futuras no treinamento.


## 2. Fundamentos matemáticos

$\mathrm{MAE}=\frac1n\sum|y_i-\hat y_i|$; $\mathrm{MSE}=\frac1n\sum(y_i-\hat y_i)^2$; $\mathrm{RMSE}=\sqrt{\mathrm{MSE}}$; $R^2=1-\frac{\sum (y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}$.

**Conexão com disciplinas anteriores:** revise os conceitos matemáticos e estatísticos correspondentes. Não repetimos integralmente os capítulos de Python, visualização e inferência estatística.


## 3. Experimento guiado: preparando a pergunta

**Antes do código**, identifique: (a) quais dados são de treino, (b) qual informação poderá entrar no modelo, (c) que métrica ou propriedade será observada, e (d) que conclusão não é permitida pelos resultados. Estes exemplos são didáticos, com semente fixa onde aplicável.


```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score
rng=np.random.default_rng(42)
X=rng.uniform(-2,2,size=(240,1));y=1+2*X[:,0]-1.5*X[:,0]**2+rng.normal(0,.45,240)
a,b,c,d=train_test_split(X,y,test_size=.25,random_state=42)
for degree in [1,2,8]:
    m=make_pipeline(PolynomialFeatures(degree=degree,include_bias=False),StandardScaler(),Ridge(alpha=1.)).fit(a,c)
    p=m.predict(b)
    print('grau',degree,'MAE',round(mean_absolute_error(d,p),3),'RMSE',round(root_mean_squared_error(d,p),3),'R2',round(r2_score(d,p),3))
```

### Como interpretar

Os números são resultados específicos do experimento e da amostra, não garantias universais. Verifique o que mudou, qual método foi usado e qual é o cenário de aplicação. Compare com a teoria matemática e não confunda desempenho de treino, validação e teste.


## 4. Segundo experimento ou extensão controlada

Mude apenas uma condição principal (algoritmo, parâmetro ou tipo de dado), quando possível, e registre o efeito. Não ajuste decisões com base no teste final.


```python
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
for name,m in [('SVR',make_pipeline(StandardScaler(),SVR(C=10,epsilon=.2))),('RandomForest',RandomForestRegressor(n_estimators=80,max_depth=6,random_state=42))]:
    m.fit(a,c);pred=m.predict(b);print(name,'MAE',round(mean_absolute_error(d,pred),3),'resíduo médio',round(float(np.mean(d-pred)),4))
```

## 5. Exercício resolvido

**Enunciado:** reforce o conceito central com um exemplo menor, cujo resultado pode ser inspecionado diretamente.


```python
# Exercício resolvido: calcular MAE, RMSE e R2 em um pequeno conjunto
from sklearn.metrics import mean_squared_error
obs=np.array([2.,4.,6.]);pred=np.array([3.,4.,5.]);print('MAE',mean_absolute_error(obs,pred),'MSE',mean_squared_error(obs,pred),'RMSE',root_mean_squared_error(obs,pred),'R2',r2_score(obs,pred))
```
